# **Matplotlib Library for Data Analytics**

#### **Histogram Charts in Matplotlib**
##### A Histogram is a graph used to show the frequency distribution of continuous numerical data.
##### It groups data into intervals called bins and shows how many values fall into each bin.
##### **Example:**
##### Student Marks:
##### [10, 20, 25, 30, 35, 40, 40, 45, 50, 55, 60]
##### Histogram shows:
| Marks Range | Frequency |
| ----------- | --------- |
| 0-20        | 2         |
| 20-40       | 5         |
| 40-60       | 4         |

##### **Used for:**
##### ✔ Data Analysis
##### ✔ Data Distribution
##### ✔ Frequency Analysis
##### ✔ Outlier Detection
##### ✔ Machine Learning Data Understanding
##### ✔ Statistical Analysis
##### **Syntax :** plt.hist(
    x,
    bins=10,
    range=None,
    density=False,
    weights=None,
    cumulative=False,
    bottom=None,
    histtype='bar',
    align='mid',
    orientation='vertical',
    rwidth=None,
    log=False,
    color=None,
    label=None,
    stacked=False,
    edgecolor=None,
    linewidth=None,
    alpha=None
)
##### **Simple Meaning :** Histogram = Data Distribution Chart
##### It answers questions like:
##### How many students scored between 40-50?
##### How many employees have salaries between 20k-30k?
##### How many customers are in each age group?
##### Data Analytics Use Cases
##### Student Marks Analysis
##### Employee Salary Analysis
##### Customer Age Distribution
##### Website Traffic Analysis
| Bar Chart        | Histogram           |
| ---------------- | ------------------- |
| Categorical Data | Numerical Data      |
| Separate Bars    | Continuous Bars     |
| Product Sales    | Marks Distribution  |
| City Revenue     | Salary Distribution |

| Parameter                | Default  | Purpose             |
| ------------------------ | -------- | ------------------- |
| `x`                      | Required | Data                |
| `bins=10`                | ✅        | Number of intervals |
| `range=None`             | ✅        | Data range          |
| `density=False`          | ✅        | Probability density |
| `cumulative=False`       | ✅        | Running frequency   |
| `histtype='bar'`         | ✅        | Histogram style     |
| `orientation='vertical'` | ✅        | Direction           |
| `color=None`             | ✅        | Bar color           |
| `edgecolor=None`         | ✅        | Border color        |
| `linewidth=None`         | ✅        | Border thickness    |
| `alpha=None`             | ✅        | Transparency        |
| `stacked=False`          | ✅        | Stacked histogram   |

#### **Exqmple :** Data 
##### salary = [25000,30000,45000,35000,28000,50000,60000,40000,42000,38000]
##### These are 10 employee salaries.
| Employee | Salary |
| -------- | ------ |
| E1       | 25000  |
| E2       | 30000  |
| E3       | 45000  |
| E4       | 35000  |
| E5       | 28000  |
| E6       | 50000  |
| E7       | 60000  |
| E8       | 40000  |
| E9       | 42000  |
| E10      | 38000  |
##### **Step 1: What does Histogram do?**
##### Histogram answers:
##### "How many salaries fall in each salary range?"
##### Instead of showing every salary individually, it groups them.
##### **Step 2: Find Min and Max**
##### Minimum Salary = 25000
##### Maximum Salary = 60000
##### Range: 60000 - 25000 = 35000
##### **Step 3: bins=5**
##### plt.hist(salary,bins=5)
##### means: Divide the data into 5 groups
##### **Step 4: Calculate Bin Width**
##### Formula: Bin Width = (Max - Min) / bins = (60000 - 25000)/5 = 35000/5 = 7000
##### So each bin covers 7000 salary units.
##### **Step 5: Create Bins**
##### Matplotlib approximately creates:
| Bin   | Range         |
| ----- | ------------- |
| Bin 1 | 25000 - 32000 |
| Bin 2 | 32000 - 39000 |
| Bin 3 | 39000 - 46000 |
| Bin 4 | 46000 - 53000 |
| Bin 5 | 53000 - 60000 |
##### **Step 6: Put Salaries Into Bins**
##### **Bin 1 :** 25000 , 28000, 30000
##### Count = 3
##### **Bin 2 :** 35000, 38000
##### Count = 2
##### **Bin 3 :** 40000, 42000, 45000
##### Count = 3
##### **Bin 4 :** 50000
##### Count = 1
##### **Bin 5 :** 60000
##### Count = 1
##### **Final Frequency Table**
| Salary Range | Frequency |
| ------------ | --------- |
| 25000-32000  | 3         |
| 32000-39000  | 2         |
| 39000-46000  | 3         |
| 46000-53000  | 1         |
| 53000-60000  | 1         |

#### **x :**
##### Input data for the histogram
##### **Syntax :** plt.hist(x)
##### **Data Type :** list, tuple, numpy array, pandas series

```python
import matplotlib.pyplot as plt
salary = [25000,30000,45000,35000,28000,50000,60000,40000,42000,38000]
plt.hist(salary)
plt.title('Salary Ranges of employees',color='red',size=20)
plt.xlabel('Salary',color='blue',size=13)
plt.ylabel('Count of Employees Salary',color='blue',size=13)
plt.xticks(color='green')
plt.yticks(color='green')
plt.grid(True,alpha=0.5)
plt.show()
```

#### **weights :**
##### The weights parameter assigns a weight (importance/value) to each data point before counting it in the histogram.
##### **Syntax :** plt.hist(x, weights=weights)
##### **Example 1 :** (Same Bin)
##### data = [40, 40, 50]
##### weights = [10, 20, 5]
##### **Without Weights**
##### plt.hist(data, bins=2)
##### Count:
##### 40 → 2 times
##### 50 → 1 time
##### **With Weights**
##### plt.hist(data, bins=2, weights=weights)
##### Matplotlib calculates:
##### 40 → 10 + 20 = 30
##### 50 → 5
##### So histogram shows:
| Value | Height |
| ----- | ------ |
| 40    | 30     |
| 50    | 5      |

```python
import matplotlib.pyplot as plt
salary = [25000,30000,45000,35000,28000,50000,60000,40000,42000,38000]
plt.hist(salary,bins=5,range=(20000,60000),weights=[2,3,4,3,2,5,6,4,4,3])
plt.title('Salary Ranges of employees',color='red',size=20)
plt.xlabel('Salary',color='blue',size=13)
plt.ylabel('Count of Employees Salary',color='blue',size=13)
plt.xticks(color='green')
plt.yticks(color='green')
plt.grid(True,alpha=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
salary = [20000,30000,40000,30000,20000,50000,60000,40000,40000,30000]
plt.hist(salary,bins=4,range=(20000,60000),weights=[2,3,4,3,2,5,6,4,4,3])
plt.title('Salary Ranges of employees',color='red',size=20)
plt.xlabel('Salary',color='blue',size=13)
plt.ylabel('Count of Employees Salary',color='blue',size=13)
plt.xticks(color='green')
plt.yticks(color='green')
plt.grid(True,alpha=0.5)
plt.show()
```

#### **cumulative :**
##### The cumulative parameter is used to show the running total (cumulative frequency) instead of the normal frequency.
##### Simple Meaning : Normally histogram shows:
##### How many values are in each bin?
##### **With:** cumulative=True
##### **Histogram shows:** How many values are in this bin + all previous bins?
##### **Example :** Data = [20000,30000,40000,30000,20000,50000,50000,40000,40000,30000]
##### 20000 = 2(total)=2 , 30000=3(total)+2(previous) = 5, 40000=3(total)+5(previous) = 8, 50000=2(total)+8 = 10

```python
import matplotlib.pyplot as plt
salary = [20000,30000,40000,30000,20000,50000,50000,40000,40000,30000]
plt.hist(salary,bins=4,cumulative=True)
plt.title('Salary Ranges of employees',color='red',size=20)
plt.xlabel('Salary',color='blue',size=13)
plt.ylabel('Count of Employees Salary',color='blue',size=13)
plt.xticks(color='green')
plt.yticks(color='green')
plt.grid(True,alpha=0.5)
plt.show()
```

#### **Multiple Histogram :**
##### Displays two or more datasets in the same graph for comparison.
##### **Syntax :** plt.hist([data1, data2])

```python
import matplotlib.pyplot as plt
IT_salary = [40000,50000,40000,60000,50000,50000,80000,20000,90000,70000]
HR_salary = [20000,30000,40000,30000,20000,50000,60000,40000,70000,30000]
plt.hist([IT_salary,HR_salary],bins=4,color=['blue','red'],alpha=0.4,label=['IT Salary','HR Salary'])
plt.title('Salary Ranges of employees',color='red',size=20)
plt.xlabel('Salary',color='blue',size=13)
plt.ylabel('Count of Employees Salary',color='blue',size=13)
plt.xticks(color='green')
plt.yticks(color='green')
plt.grid(True,alpha=0.5)
plt.legend()
plt.show()
```
