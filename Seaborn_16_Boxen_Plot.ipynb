# **Seaborn Library for Data Analytics**

#### **What is a Boxen Plot :**
##### A Boxen Plot shows the distribution of numerical data using multiple nested boxes.
##### It provides more information about the tails (extreme values) than a normal Box Plot.
##### **Definition :** A Boxen Plot visualizes the distribution of data with multiple quantile-based boxes, making it especially useful for large datasets.
##### **Why Use Boxen Plot :**
##### Better than Box Plot for large datasets
##### Shows data tails clearly
##### Detects outliers
##### Shows distribution shape
##### Useful in EDA
##### Used in Data Science & ML projects
##### **Syntax :** sns.boxenplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    order=None,
    hue_order=None,
    orient=None,
    color=None,
    palette=None,
    saturation=0.75,
    width=0.8,
    dodge='auto',
    linewidth=None,
    linecolor=None,
    k_depth='tukey',
    outlier_prop=0.007,
    trust_alpha=0.05,
    showfliers=True,
    fill=True,
    gap=0,
    legend='auto'
)
| Box Plot                 | Boxen Plot                |
| ------------------------ | ------------------------- |
| Shows Q1, Q2, Q3         | Shows Multiple Quantiles  |
| Less Detail              | More Detail               |
| Small Data               | Large Data                |
| Basic Distribution       | Advanced Distribution     |
| Simple Outlier Detection | Better Tail Visualization |


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
df=pd.read_csv('Employee_Dataset.csv')
df
```

##### **X :** Categorical Variable
##### **Y :** Numerical Variable

```python
sns.boxenplot(data=df,x='Department',y='Experience_Years')
plt.title('Boxen Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Experience',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.5)
plt.tight_layout()
plt.show()
```
