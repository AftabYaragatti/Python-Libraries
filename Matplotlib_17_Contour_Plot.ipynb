# **Matplotlib Library for Data Analytics**

### **Contour Plot in Matplotlib**

##### A Contour Plot is used to visualize 3D data on a 2D graph using lines or colors.
##### It is also called an Isoline Plot because it connects points having the same Z value.
##### **1. What is a Contour Plot :**
##### Suppose you have:
##### X Coordinate
##### Y Coordinate
##### Height (Z)
##### This is 3D data.
##### **Example:**
| X | Y | Height(Z) |
| - | - | --------- |
| 1 | 1 | 100       |
| 2 | 1 | 100       |
| 3 | 1 | 100       |
| 1 | 2 | 200       |
| 2 | 2 | 200       |
##### Instead of creating a 3D graph, Contour Plot shows equal-height regions using contour lines.
##### **Why Use Contour Plot?**
##### Used when you have: X, Y, Z and want to understand:Peaks, Valleys, Density, Temperature Distribution, Pressure Distribution & Elevation Maps
##### **Types of Contour Plots :**
| Plot                | Function         |
| ------------------- | ---------------- |
| Contour Plot        | `plt.contour()`  |
| Filled Contour Plot | `plt.contourf()` |
##### **Syntax :** plt.contour(
    X,
    Y,
    Z,
    levels=10,
    colors=None,
    cmap=None,
    linewidths=1,
    linestyles='solid'
)
##### Understanding X, Y and Z
##### A contour plot requires:
##### X → Horizontal values
##### Y → Vertical values
##### Z → Height / Intensity values

##### **Creating simple data :**
##### What is meshgrid() : Creates coordinate pairs.
##### X,Y=np.meshgrid(x,y)

```python
import numpy as np
x=np.linspace(-5,5,100)
y=np.linspace(-5,5,100)
X,Y=np.meshgrid(x,y)
Z=X**2 + Y**2
print(X,Y,Z)
```

##### **Z must be 2D or 3D :**

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(-5,5,100)
y=np.linspace(-5,5,100)
X,Y=np.meshgrid(x,y)
Z=X**2 + Y**2
plt.contour(X,Y,Z)
plt.title("Contour Plot of Z = X² + Y²")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.show()
```

#### **levels :**
##### Controls number of contour lines.
##### Default : Matplotlib chooses automatically.

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(-5,5,100)
y=np.linspace(-5,5,100)
X,Y=np.meshgrid(x,y)
Z=X**2 + Y**2
plt.contour(X,Y,Z,levels=15)
plt.title("Contour Plot of Z = X² + Y²")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.show()
```

#### **colors :**

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(-5,5,100)
y=np.linspace(-5,5,100)
X,Y=np.meshgrid(x,y)
Z=X**2 + Y**2
plt.contour(X,Y,Z,levels=5,colors=['r','g','b'])
plt.title("Contour Plot of Z = X² + Y²")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.show()
```

#### **cmap :**
##### Colormap. Used mostly with filled contours.
| Value   | Theme             |
| ------- | ----------------- |
| viridis | Blue-Green-Yellow |
| plasma  | Purple-Yellow     |
| inferno | Dark-Yellow       |
| magma   | Dark-Pink         |
| cool    | Cyan-Magenta      |
| hot     | Black-Red-Yellow  |

##### **Note :** Either colors or cmap must be None

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(-5,5,100)
y=np.linspace(-5,5,100)
X,Y=np.meshgrid(x,y)
Z=X**2 + Y**2
plt.contour(X,Y,Z,levels=5,colors=None,cmap='hot')
plt.title("Contour Plot of Z = X² + Y²")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.show()
```

#### **linewidths & linestyles :**

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(-5,5,100)
y=np.linspace(-5,5,100)
X,Y=np.meshgrid(x,y)
Z=X**2 + Y**2
plt.contour(X,Y,Z,levels=5,colors=None,cmap='hot',linewidths=2,linstyles='solid')
plt.title("Contour Plot of Z = X² + Y²")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.show()
```

| Contour Plot         | Scatter Plot            |
| -------------------- | ----------------------- |
| Uses X,Y,Z           | Uses X,Y                |
| Shows density/height | Shows points            |
| Continuous surface   | Individual observations |

| Contour Plot       | Heatmap              |
| ------------------ | -------------------- |
| Uses contour lines | Uses colored cells   |
| Smooth transitions | Block representation |

| Contour Plot   | Surface Plot   |
| -------------- | -------------- |
| 2D View        | 3D View        |
| Easier to read | More realistic |

### **Filled Contour Plot(Contourf Plot) in Matplotlib**

##### **Syntax :** plt.contourf(
    X,
    Y,
    Z,
    levels=10,
    cmap='viridis'
    colors=None,
    linewidths=1,
    linestyles='solid'
)

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(-5,5,100)
y=np.linspace(-5,5,100)
X,Y=np.meshgrid(x,y)
Z=X**2 + Y**2
plt.contourf(X,Y,Z,levels=5,colors=None,cmap='hot',linewidths=2,linstyles='solid')
plt.title("Contour Plot of Z = X² + Y²")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(-5,5,100)
y=np.linspace(-5,5,100)
X,Y=np.meshgrid(x,y)
Z=X**2 + Y**2
plt.contourf(X,Y,Z,levels=5,colors=None,cmap='cool',linewidths=2,linstyles='solid')
plt.title("Contour Plot of Z = X² + Y²")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.show()
```

#### **plt.colorbar(label="Z Value")**

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(-5,5,100)
y=np.linspace(-5,5,100)
X,Y=np.meshgrid(x,y)
Z=X**2 + Y**2
plt.contourf(X,Y,Z,levels=5,colors=None,cmap='plasma',linewidths=2,linstyles='solid')
plt.title("Contour Plot of Z = X² + Y²")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.colorbar(label="Z Value")
plt.show()
```
