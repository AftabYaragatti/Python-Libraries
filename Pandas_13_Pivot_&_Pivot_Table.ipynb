# **Pandas Library**
### **Pivot & Pivot Table**

##### **Pivot :**
##### It is used to reshape the existing  dataframe. If we have any duplicates it doesn't work. In pivot we can't use aggregate functions.
##### **Syntax : pd.pivot(data=DataFrame,index='row_column',columns='column_to_expand',values='value_column')**
##### Here for index & columns we want to provide only categorical columns.

| Parameter | Meaning                                            |
| --------- | -------------------------------------------------- |
| `data`    | The DataFrame to reshape                           |
| `index`   | Column whose values become row labels              |
| `columns` | Column whose unique values become new column names |
| `values`  | Column whose values fill the new table             |


```python
import pandas as pd
df = pd.DataFrame({
    'Name': ['Aftab', 'Aftab', 'John', 'John'],
    'Subject': ['Math', 'Science', 'Math', 'Science'],
    'Marks': [90, 85, 80, 75]
})
df
```

```python
pivot_df = pd.pivot(data=df,index='Name',columns='Subject',values='Marks')
pivot_df
```

```python
pivot_df = pd.pivot(data=df,index='Name',columns='Subject')
pivot_df
```

#### **Before Pivot :**
| Name  | Subject | Marks |
| ----- | ------- | ----- |
| Aftab | Math    | 90    |
| Aftab | Science | 85    |
| John  | Math    | 80    |
| John  | Science | 75    |
#### **After Pivot :**
| Name  | Math | Science |
| ----- | ---- | ------- |
| Aftab | 90   | 85      |
| John  | 80   | 75      |

##### **IMPORTANT NOTE :**
##### pivot() requires that each combination of index and columns is unique.
##### If duplicate combinations exist:
##### ValueError: Index contains duplicate entries, cannot reshape
##### Use pivot_table() instead:
##### **Example : pd.pivot_table(data=df,index='Name',columns='Subject',values='Marks',aggfunc='mean')**

```python
import pandas as pd
import numpy as np
df=pd.read_csv("Employee_Cleaned_Dataset.csv")
df.drop(columns='Unnamed: 0', inplace=True)
df
```

##### **Here we will get error because index should not conain any duplicate values**

```python
pivot_df = pd.pivot(data=df,index='City',columns='Department',values='Salary')
pivot_df
```

```python
pivot_df = pd.pivot(data=df,index='Employee ID',columns='Department',values='Salary')
pivot_df
```

##### **Suppose if we are not specified the values it will take first unused column from the data set as values**
##### **Example :** We have used Emp ID as index, Department as Columns and values are not specified, at that time the interpreter itself takes the first unused column from the dataset i,.e Employee Name.

```python
pivot_df = pd.pivot(data=df,index='Employee ID',columns='Department')
pivot_df
```

##### **Pivot Table :**
##### It is used to perform reshaping of existing dataframe along with aggragation function. It accepts duplicate values.
##### **Syntax : pd.pivot_table(data=DataFrame,index='categorical_column',aggfunc=['fun_name','fun_name'],fill_value=value,margins=True)**
##### **Note :** Margins=True works with count aggragation function or when we want to perform count aggregation then we use margins=True.

```python
pd.pivot_table(data=df,index='Name',columns='Subject',values='Marks',aggfunc='mean')
```

```python
df = pd.DataFrame({
    'Name': ['Aftab', 'Aftab', 'John', 'John'],
    'Subject': ['Math', 'Math', 'Science', 'Science'],
    'Marks': [90, 80, 70, 60]
})
```

```python
df
```

##### **Example 1: Mean of values**

```python
pd.pivot_table(
    data=df,
    index='Name',
    columns='Subject',
    values='Marks',
    aggfunc='mean'
)
```

##### **Example 2: Sum of values**

```python
pd.pivot_table(
    data=df,
    index='Name',
    columns='Subject',
    values='Marks',
    aggfunc='sum'
)
```

##### **Example 3: Count of values**

```python
pd.pivot_table(
    data=df,
    index='Name',
    columns='Subject',
    values='Marks',
    aggfunc='count'
)
```

##### **Example 4: Multiple Aggregation of values**

```python
pd.pivot_table(
    data=df,
    index='Name',
    columns='Subject',
    values='Marks',
    aggfunc=['sum', 'mean', 'max']
)
```

##### **Example 5: Fill Null values**

```python
pd.pivot_table(
    data=df,
    index='Name',
    columns='Subject',
    values='Marks',
    aggfunc='mean',
    fill_value=0
)
```

```python
import pandas as pd
import numpy as np
df=pd.read_csv("Employee_Cleaned_Dataset.csv")
df.drop(columns='Unnamed: 0', inplace=True)
df
```

```python
df.columns = df.columns.str.strip()
```

```python
pd.pivot_table(
    data=df,
    index='Department',
    values='Salary',
    aggfunc='mean'
)
```

```python
pd.pivot_table(
    data=df,
    index='Department',
    values='Salary',
    aggfunc='sum'
)
```

```python
pd.pivot_table(
    data=df,
    index='Department',
    values='Salary',
    aggfunc=['sum', 'mean', 'max','min']
)
```

```python
pd.pivot_table(
    data=df,
    index='Department',
    columns='Employee ID',
    values='Salary',
    aggfunc=['sum', 'mean', 'max','min']
)
```

```python
pd.pivot_table(
    data=df,
    index='Department',
    columns='Employee ID',
    values='Salary',
    aggfunc=['sum', 'mean', 'max','min'],
    fill_value=0
)
```

###### **pivot() vs pivot_table()**
| Feature                         | `pivot()` | `pivot_table()` |
| ------------------------------- | --------- | --------------- |
| Handles duplicates              | ❌ No      | ✅ Yes           |
| Aggregation                     | ❌ No      | ✅ Yes           |
| Functions like mean, sum, count | ❌ No      | ✅ Yes           |
| More flexible                   | ❌ No      | ✅ Yes           |
