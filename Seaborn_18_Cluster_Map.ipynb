# **Seaborn Library for Data Analytics**

#### **What is a Cluster Plot :**
##### A Cluster Plot (Clustermap) is an advanced Heatmap that:
##### Displays data as a Heatmap
##### Automatically groups similar rows and columns
##### Uses Hierarchical Clustering
##### **Definition :** A Clustermap visualizes a matrix dataset as a heatmap while grouping similar rows and columns using hierarchical clustering.
##### **Why Use Cluster Plot :**
##### Correlation Analysis
##### Customer Segmentation
##### Gene Expression Analysis
##### Pattern Detection
##### Feature Grouping
##### Machine Learning EDA
##### Data Mining
##### **Difference Between Heatmap and Clustermap :**
Heatmap	Clustermap
| Heatmap                   | Clustermap              |
| ------------------------- | ----------------------- |
| Only color representation | Color + Clustering      |
| No grouping               | Automatic grouping      |
| Simpler                   | More advanced           |
| Fixed order               | Reordered automatically |

##### **Example :**
Heatmap
A  B  C  D
1  2  8  9
2  1  9  8
8  9  1  2
9  8  2  1
##### **Clustermap :**
A  B | C  D
1  2 | 8  9
2  1 | 9  8
--------------
8  9 | 1  2
9  8 | 2  1
##### Similar rows/columns are grouped automatically.
##### **Hierarchical Clustering :**
##### **Cluster Plot uses:**
##### Agglomerative Clustering
##### **Step 1 :** Each point = Separate cluster
##### **Step 2 :**Nearest clusters merged
##### **Step 3 :** Repeat
##### **Step 4 :**Final tree (Dendrogram)
##### **What is a Dendrogram :** A tree-like diagram showing similarity between observations.
##### Closer branches = More similar.
##### **Syntax :** sns.clustermap(
    data,
    pivot_kws=None,
    method='average',
    metric='euclidean',
    z_score=None,
    standard_scale=None,
    figsize=(10,10),
    cbar_kws=None,
    row_cluster=True,
    col_cluster=True,
    row_colors=None,
    col_colors=None,
    dendrogram_ratio=0.2,
    colors_ratio=0.03,
    cbar_pos=(0.02,0.8,0.05,0.18),
    tree_kws=None,
    cmap=None,
    center=None,
    annot=False,
    linewidths=0,
    linecolor='white'
)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
df=pd.read_csv('Employee_Dataset.csv')
df
```

#### **data :**
##### Dataset or matrix.
##### **Example :** data=df.corr()

```python
numeric_df = df.select_dtypes(include=['number'])
corr = df.corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.clustermap(corr,cmap='coolwarm',annot=True)
plt.title("Correlation Clustermap")
plt.show()
```
