### **Scatter Plot :**
#### **What is a Scatter Plot :**
##### A Scatter Plot is used to visualize the relationship between two numerical variables.
##### Each point represents one observation (row) in the dataset.
##### **Example :**
| Age | Salary |
| --- | ------ |
| 21  | 25000  |
| 22  | 30000  |
| 23  | 35000  |
| 24  | 40000  |
##### Every row becomes one dot on the graph.
##### **Why Use Scatter Plot :**
##### Find correlation
##### Find trends
##### Detect outliers
##### Identify clusters
##### Compare variables
##### Data Exploration (EDA)
##### Machine Learning preprocessing

##### **Complete Syntax :** sns.scatterplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    style=None,
    size=None,
    palette=None,
    sizes=None,
    markers=True,
    alpha=None,
    legend='auto',
    color=None,
    marker='o'
)

#### **legend :**
##### Controls legend display.
##### **Values**
##### legend='auto'
##### legend='brief'
##### legend='full'
##### legend=False
##### **legend='auto' (Default)**
##### Seaborn automatically decides how much information to show.
##### For a small number of categories → shows all categories.
##### For many categories → may simplify the legend.
##### **legend='brief'**
##### Shows a summarized legend.
##### Mainly useful when the hue variable is numeric and contains many unique values.
##### **legend='full'**
##### Displays every unique value in the legend.
##### legend=False : Removes the legend completely.

| Value | Meaning                           | Best For           |
| ----- | --------------------------------- | ------------------ |
| auto  | Seaborn decides                   | General use        |
| brief | Shows fewer representative values | Large numeric data |
| full  | Shows all values                  | Detailed analysis  |
| False | No legend                         | Clean plots        |


```python
sns.scatterplot(data=df,x='Degree Yop',y='Degree Percentage', hue='Courses Taken',size='Admission Type',sizes=(35,100),marker='D',legend='auto')
plt.title('Scatter Plot (Degree YOP VS PERCENTAGE',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.xlabel('Degree YOP',color='purple',fontsize=12)
plt.ylabel('Degree Percentage',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True)
plt.legend(loc='best',bbox_to_anchor=(1.05,1))
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=df,x='Degree Yop',y='Degree Percentage', hue='Courses Taken',size='Admission Type',sizes=(35,100),marker='D',legend='full')
plt.title('Scatter Plot (Degree YOP VS PERCENTAGE',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.xlabel('Degree YOP',color='purple',fontsize=12)
plt.ylabel('Degree Percentage',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True)
plt.legend(loc='best',bbox_to_anchor=(1.05,1))
plt.tight_layout()
plt.show()
```

#### **BEST EXAMPLE**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df=pd.DataFrame({
    'Age':[21,22,23,24,25],
    'Salary':[25000,30000,35000,40000,45000],
    'Gender':['Male','Female','Male','Female','Male'],
    'Experience':[1,2,3,4,5],
    'Department':['IT','HR','Sales','IT','HR']
})

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x='Age',
    y='Salary',
    hue='Gender',
    style='Department',
    size='Experience',
    palette='Set2',
    sizes=(100,500),
    alpha=1,
    marker='o',
    legend='full'
)

plt.title("Employee Salary Analysis")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid(True)

plt.show()
```
