# **Matplotlib Library for Data Analytics**

#### **3D Plotting in Matplotlib**
##### **What is 3D Plotting :** 3D plotting is used to visualize data in three dimensions:
##### X-axis → Width
##### Y-axis → Height
##### Z-axis → Depth
##### Unlike normal plots (2D), 3D plots can represent an additional variable
##### **Required Import**
##### import matplotlib.pyplot as plt
##### from mpl_toolkits.mplot3d import Axes3D
##### Axes3D enables 3D plotting functionality.
##### **Basic Syntax :**
##### fig = plt.figure()
##### ax = fig.add_subplot(111, projection='3d')
##### plt.show()
##### **Meaning of 111**
##### 1 row
##### 1 column
##### 1st plot
##### **Creating a 3D Axes :**
##### Modern Method
##### fig = plt.figure()
##### ax = plt.axes(projection='3d')
##### **OR**
##### fig = plt.figure()
##### ax = fig.add_subplot(111,projection='3d')

##### **Common 3D Plot types :**
| Plot               | Function            |
| ------------------ | ------------------- |
| 3D Line Plot       | ax.plot3D()         |
| 3D Scatter Plot    | ax.scatter3D()      |
| 3D Surface Plot    | ax.plot_surface()   |
| 3D Wireframe Plot  | ax.plot_wireframe() |
| 3D Contour Plot    | ax.contour3D()      |
| 3D Bar Plot        | ax.bar3d()          |
| 3D Stem Plot       | ax.stem()           |
| 3D Trisurface Plot | ax.plot_trisurf()   |

#### **1. 3D Line Plot :**
##### Shows a line in 3D space.
##### **Syntax :** ax.plot3D(
    xs,
    ys,
    zs,
    color=None,
    linewidth=1,
    linestyle='-',
    label=None
)

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
x=[1,2,3,4,5]
y=[7,8,9,10,11]
z=[13,15,17,18,19]
fig=plt.figure()
ax=fig.add_subplot(111,projection='3d')
ax.plot3D(x,y,z)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
# Create figure
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')
# Data
x = np.linspace(0, 10, 100)
y = np.sin(x)
z = np.cos(x)
# 3D Line
ax.plot(x, y, z)
# Labels
ax.set_title("3D Line Plot")
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")
plt.show()
```

#### **3. 3D Bar Plot :**
##### Displays bars in 3D.
##### **Syntax :** ax.bar3d(
    x,
    y,
    z,
    dx,
    dy,
    dz,
    color=None,
    alpha=1
)
| Parameter | Meaning           |
| --------- | ----------------- |
| x,y,z     | Starting position |
| dx        | Width             |
| dy        | Depth             |
| dz        | Height            |

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
# Create figure
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')
# X and Y positions
x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 1, 2, 3, 4])
# Starting position on Z-axis
z = np.zeros(5)
# Width, Depth, Height
dx = np.ones(5) * 0.5
dy = np.ones(5) * 0.5
dz = np.array([10, 25, 15, 30, 20])
# 3D Bars
ax.bar3d(x, y, z, dx, dy, dz)
# Labels
ax.set_title("3D Bar Plot")
ax.set_xlabel("Products")
ax.set_ylabel("Stores")
ax.set_zlabel("Sales")
plt.show()
```
