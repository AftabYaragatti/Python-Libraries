# **Pandas Library**
### **Attributes in Pandas DataFrame()**

##### 1. df.shape : Returns number of rows and columns
##### 2. df.size : Total number of elements
##### 3. df.ndim : Number of dimensions, Always 2 for DataFrame
##### 4. df.index : Row labels
##### 5. df.columns : Column labels
##### 6. df.dtypes : Data types of each column
##### 7. df.values : Returns data as NumPy array(Now prefer to_numpy())
##### 8. df.axes : List of index + columns
##### 9. df.empty : Checks if DataFrame is empty
##### 10. df.T : Transpose (rows ↔ columns)
##### 11. df.attrs : Custom metadata dictionary Syntax : df.attrs['source'] = 'HR Dataset'
##### 12. df.shape[0] : Number of rows

#### **WHAT ARE ATTRIBUTES?**
##### Attributes are **properties (no brackets `()`)** used to **inspect DataFrame structure and metadata**
##### df.shape   ✔ attribute
##### df.head()  ❌ method

##### **1. df.shape :** Returns number of rows and columns in the form of tuple.
