# **Matplotlib Library for Data Analytics**

### **Hexbin plot in Matplotlib**

##### A Hexbin Plot is an advanced version of a Scatter Plot used when you have thousands or millions of data points.(Many points overlap.)
##### Instead of plotting every point, it groups points into hexagonal bins and colors them according to density.
##### **Each hexagon shows:** Number of Points inside the Hexagon
##### Suppose you have: 100 points - Scatter Plot works fine.
##### Suppose you have: 10000 points - Scatter Plot becomes impossible to understand.
##### **Hexbin Plot solves this problem by showing:**
##### High Density Area
##### Medium Density Area
##### Low Density Area
##### **Syntax :** plt.hexbin(
    x,
    y,
    C=None,
    gridsize=100,
    bins=None,
    xscale='linear',
    yscale='linear',
    extent=None,
    cmap=None,
    mincnt=None,
    marginals=False,
    alpha=None,
    linewidths=None
)

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.random.randn(500)
y=np.random.randn(500)
plt.hexbin(x,y)
plt.show()
```

#### **C :**
##### Additional values used for aggregation.

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.random.randn(500)
y=np.random.randn(500)
sales=np.random.randn(500)
plt.hexbin(x,y,C=sales)
plt.show()
```

#### **gridsize :**
##### Controls number of hexagons
##### gridsize=50 - Many small hexagons.
##### gridsize=10 - Few large hexagons.
##### gridsize=10
##### ⬢ ⬢ ⬢
##### gridsize=50
##### ⬢⬢⬢⬢⬢⬢⬢⬢⬢⬢

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.random.randn(500)
y=np.random.randn(500)
sales=np.random.randn(500)
plt.hexbin(x,y,C=sales,gridsize=70)
plt.show()
```

#### **bins :**
##### Controls color scaling.
| Value   | Meaning       |
| ------- | ------------- |
| None    | Linear counts |
| 'log'   | Logarithmic   |
| integer | Fixed bins    |

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.random.randn(500)
y=np.random.randn(500)
sales=np.random.randn(500)
plt.hexbin(x,y,C=sales,gridsize=70,bins='log')
plt.show()
```

#### **extent :**
##### Controls plotting region.
##### **Syntax :** extent=[xmin,xmax,ymin,ymax]
| Range    | Percentage of Data |
| -------- | ------------------ |
| -1 to +1 | 68%                |
| -2 to +2 | 95%                |
| -3 to +3 | 99.7%              |

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.random.randn(5000)
y=np.random.randn(5000)
sales=np.random.randn(5000)
plt.hexbin(x,y,C=sales,gridsize=70,extent=[-3,3,-3,3])
plt.show()
```

#### **cmap :**
##### Color Map
| Value   | Color Theme       |
| ------- | ----------------- |
| viridis | Blue-Green-Yellow |
| plasma  | Purple-Yellow     |
| inferno | Dark-Yellow       |
| magma   | Dark-Pink         |
| cool    | Cyan-Magenta      |
| hot     | Black-Red-Yellow  |


```python
import matplotlib.pyplot as plt
import numpy as np
x=np.random.randn(500)
y=np.random.randn(500)
sales=np.random.randn(500)
plt.hexbin(x,y,C=sales,gridsize=70,extent=[-3,3,-3,3],cmap='cool')
plt.show()
```

#### **mincnt :**
##### Minimum count required. mincnt=5 - Hide hexagons containing fewer than 5 points.

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.random.randn(500)
y=np.random.randn(500)
sales=np.random.randn(500)
plt.hexbin(x,y,C=sales,gridsize=70,extent=[-3,3,-3,3],cmap='cool',mincnt=2)
plt.show()
```

#### **marginals :**
##### Adds histograms.
##### Histogram X -> Hexbin Plot -> Histogram Y

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.random.randn(500)
y=np.random.randn(500)
sales=np.random.randn(500)
plt.hexbin(x,y,C=sales,gridsize=70,extent=[-3,3,-3,3],cmap='cool',mincnt=1)
plt.show()
```

#### **alpha :**
##### Transparency

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.random.randn(500)
y=np.random.randn(500)
sales=np.random.randn(500)
plt.hexbin(x,y,C=sales,gridsize=70,extent=[-3,3,-3,3],cmap='cool',mincnt=1,alpha=0.6)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
sales_2024=[84,45,55,51,72,62,81,69,92,37,53,99]
sales_2025=[94,51,56,59,75,69,84,78,97,34,54,98]
plt.hexbin(sales_2024,sales_2025,gridsize=5,cmap='viridis')
plt.colorbar(label='Count')
plt.title('Sales 2024 vs Sales 2025')
plt.xlabel('Sales 2024')
plt.ylabel('Sales 2025')
plt.show()
```

| Hexbin Plot           | Scatter Plot        |
| --------------------- | ------------------- |
| Large datasets        | Small datasets      |
| Density visualization | Individual points   |
| No overlap issue      | Overlapping points  |
| Faster for big data   | Slower for big data |
    
| Hexbin Plot       | Histogram              |
| ----------------- | ---------------------- |
| 2D data           | 1D data                |
| X and Y variables | Single variable        |
| Density map       | Frequency distribution |
