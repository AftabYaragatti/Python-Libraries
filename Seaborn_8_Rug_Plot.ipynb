# **Seaborn Library for Data Analytics**

#### **What is a Rug Plot :**
##### A Rug Plot is a simple visualization that displays individual data points as small lines (ticks) on an axis.
##### Each tick represents one observation in the dataset.
##### **Definition :** A Rug Plot shows the distribution of data by drawing a small line for every data value

#### **Why Use Rug Plot :**
##### Visualize individual observations
##### Identify data concentration
##### Detect gaps in data
##### Detect outliers
##### Support Histogram
##### Support KDE Plot
##### Exploratory Data Analysis (EDA)
##### **Syntax :** sns.rugplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    height=0.05,
    expand_margins=True,
    palette=None,
    color=None,
    alpha=None,
    linewidth=1,
    legend=True
)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('Employee_Dataset.csv')
df
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.rugplot(data=df,x='Experience_Years')
plt.title('Rug Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.xlabel('Years of Experience',color='purple',fontsize=12)
plt.ylabel('')
plt.xticks(color='blue',rotation=0)
plt.yticks([])
plt.grid(True,axis='both',alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.rugplot(data=df,y='Experience_Years')
plt.title('Rug Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Years of Experience',color='purple',fontsize=12)
plt.xlabel('')
plt.yticks(color='blue',rotation=0)
plt.xticks([])
plt.grid(True,axis='both',alpha=0.2)
plt.tight_layout()
plt.show()
```

#### **hue :**
##### Creates separate distributions using colors.
##### **Example :** hue='Gender'

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.rugplot(data=df,x='Experience_Years',hue='Department')
plt.title('Rug Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.xlabel('Years of Experience',color='purple',fontsize=12)
plt.ylabel('')
plt.xticks(color='blue',rotation=0)
plt.yticks([])
plt.grid(True,axis='both',alpha=0.4)
plt.tight_layout()
plt.show()
```

#### **height :**
##### Controls tick length.
| Value | Result       |
| ----- | ------------ |
| 0.02  | Small ticks  |
| 0.05  | Default      |
| 0.1   | Medium ticks |
| 0.2   | Long ticks   |

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.rugplot(data=df,x='Experience_Years',hue='Department',height=0.5)
plt.title('Rug Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.xlabel('Years of Experience',color='purple',fontsize=12)
plt.ylabel('')
plt.xticks(color='blue',rotation=0)
plt.yticks([])
plt.grid(True,axis='both',alpha=0.4)
plt.tight_layout()
plt.show()
```

#### **expand_margins :**
