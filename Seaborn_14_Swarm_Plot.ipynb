# **Seaborn Library for Data Analytics**

#### **What is a Swarm Plot :**
##### A Swarm Plot is used to display individual data points for different categories without overlapping.
##### It is an improved version of the Strip Plot.
##### **Definition :** A Swarm Plot visualizes all observations while automatically adjusting points so that they do not overlap.
##### **Why Use Swarm Plot :**
##### Visualize Every Observation
##### Compare Categories
##### Detect Outliers
##### Understand Distribution
##### Exploratory Data Analysis (EDA)
##### Statistical Analysis
##### Machine Learning Data Exploration
##### **Example Data :**
| Department | Salary |
| ---------- | ------ |
| IT         | 50000  |
| IT         | 51000  |
| IT         | 52000  |
| HR         | 35000  |
| HR         | 37000  |
| Sales      | 45000  |
    
| Swarm Plot                      | Strip Plot                |
| ------------------------------- | ------------------------- |
| No overlap                      | May overlap               |
| Cleaner visualization           | Simpler                   |
| Better for small datasets       | Better for large datasets |
| Automatically adjusts positions | Uses random jitter        |
##### **Syntax :** sns.swarmplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    order=None,
    hue_order=None,
    dodge=False,
    orient=None,
    color=None,
    palette=None,
    size=5,
    alpha=None,
    linewidth=0,
    edgecolor=None,
    legend='auto'
)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
df=pd.read_csv('Employee_Dataset.csv')
df
```

##### **X :** Categorical Variable
##### **Y :** Numerical Variable

```python
sns.swarmplot(data=df,x='Department',y='Experience_Years',hue='Gender')
plt.title('Swarm Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Experience',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
sns.stripplot(data=df,x='Department',y='Experience_Years',hue='Gender')
plt.title('Strip Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Experience',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.5)
plt.tight_layout()
plt.show()
```
