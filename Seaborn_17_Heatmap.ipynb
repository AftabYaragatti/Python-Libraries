# **Seaborn Library for Data Analytics**

#### **What is a Heatmap :**
##### A Heatmap is a graphical representation of data where values are represented by colors.
##### Higher values and lower values are shown using different color intensities.
##### **Definition :** A Heatmap visualizes numerical data in a matrix format using colors to represent values.
##### **Why is it called a Heatmap :** Because values are represented as heat (color intensity).
##### High Value  → Dark/Hot Color 
##### Medium Value → Medium Color
##### Low Value → Light/Cool Color 
##### **Why Use Heatmap :**
##### Correlation Analysis
##### Feature Selection
##### Data Pattern Detection
##### Missing Value Analysis
##### Business Intelligence
##### Machine Learning EDA
##### Statistical Analysis
##### **Real-World Applications :**
##### **Data Analytics :** Find relationships between variables.
##### **Machine Learning :** Identify highly correlated features.
##### **Finance :** Stock correlation analysis.
##### **Sales Analysis :** Region-wise performance.
##### **Healthcare :** Patient data visualization.

##### **Syntax :** sns.heatmap(
    data,
    vmin=None,
    vmax=None,
    cmap=None,
    center=None,
    annot=False,
    fmt='.2g',
    annot_kws=None,
    linewidths=0,
    linecolor='white',
    cbar=True,
    cbar_kws=None,
    square=False,
    xticklabels='auto',
    yticklabels='auto',
    mask=None
)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
df=pd.read_csv('Employee_Dataset.csv')
df
```

#### **data :**
##### Dataset or matrix to visualize.
##### **Example :** data=df.corr()

#### **Heatmap :** Only works or deals with numerical values only....

#### **vmin :** 
##### Minimum value for color scale.
##### **Example :** vmin=0

```python
numeric_df = df.select_dtypes(include=['number'])
corr = df.corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr,vmin=0)
plt.title("Correlation Heatmap")
plt.show()
```

##### **vmax :** Maximum value for color scale.
##### **Example :** vmax=1

```python
numeric_df = df.select_dtypes(include=['number'])
corr = df.corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr,vmin=0,vmax=3)
plt.title("Correlation Heatmap")
plt.show()
```

##### **cmap :**
##### Controls color palette.
##### **Values :** viridis , plasma , magma , inferno , coolwarm , Spectral , RdBu , YlGnBu , Blues , Greens , Reds , Purples , Oranges

```python
numeric_df = df.select_dtypes(include=['number'])
corr = df.corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

#### **center :**
##### Center value for diverging colormaps.
##### **Example :** center=0
##### Used frequently for correlation matrices.

```python
numeric_df = df.select_dtypes(include=['number'])
corr = df.corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr, cmap='coolwarm',center=0)
plt.title("Correlation Heatmap")
plt.show()
```

#### **mask :**
##### Hide selected cells.
##### **Example :** mask=np.triu(df.corr())
##### Used for upper triangle hiding

```python
import numpy as np
numeric_df = df.select_dtypes(include=['number'])
corr = df.corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr, cmap='coolwarm',annot=True,mask=np.triu(numeric_df.corr()))
plt.title("Correlation Heatmap")
plt.show()
```
