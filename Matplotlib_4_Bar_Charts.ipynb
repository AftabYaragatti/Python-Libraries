# **Matplotlib Library for Data Analytics**

### **Bar Charts in Matplotlib**
##### A Bar Chart is used to compare values between different categories.
##### **What is a Bar Chart?**
##### A Bar Chart is a graph used to compare values among different categories using rectangular bars. The height (or length) of each bar represents the value of that category.
##### **Example :** Employees and their salaries:
##### Employee	Salary
##### A	25000
##### B	30000
##### C	45000
| Use Case               | X-Axis      | Y-Axis  |
| ---------------------- | ----------- | ------- |
| Department-wise Salary | Departments | Salary  |
| Product-wise Sales     | Products    | Sales   |
| City-wise Revenue      | Cities      | Revenue |
| Student Marks          | Students    | Marks   |
| Company Profit         | Companies   | Profit  |
| Employee Performance   | Employees   | Score   |
##### **Uses of Bar Charts :**
| Use Case           | Example                |
| ------------------ | ---------------------- |
| Compare Categories | Product A vs Product B |
| Compare Sales      | Product-wise Sales     |
| Compare Revenue    | City-wise Revenue      |
| Compare Salary     | Employee Salary        |
| Compare Marks      | Student Marks          |
| Compare Population | City Population        |
| Compare Profit     | Company Profit         |
| Compare Expenses   | Department Expenses    |
##### **Syntax :** plt.bar(x, height, width=0.8, bottom=None, align='center', color=None, edgecolor=None, linewidth=None, tick_label=None, alpha=None, label=None, hatch=None, zorder=None)
| Parameter  | Default Value | Description       |
| ---------- | ------------- | ----------------- |
| x          | Required      | Categories        |
| height     | Required      | Bar heights       |
| width      | 0.8           | Width of bars     |
| bottom     | None          | Starting position |
| align      | 'center'      | Bar alignment     |
| color      | None          | Fill color        |
| edgecolor  | None          | Border color      |
| linewidth  | None          | Border thickness  |
| tick_label | None          | Custom labels     |
| alpha      | None          | Transparency      |
| label      | None          | Legend label      |
| hatch      | None          | Bar pattern       |
| zorder     | 0             | Layer order       |

```python
import matplotlib.pyplot as plt
years = [2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024,2025]
growth = [15,19,22,26,28,29,32,35,38,34,21,22,28,34,38,67]
plt.bar(years, growth)
plt.show() 
```

#### **1. Vertical bar Charts :**
##### A Vertical Bar Chart is used to compare values across different categories using vertical rectangular bars.
##### Categories are displayed on the X-axis
##### Values are displayed on the Y-axis
##### Height of each bar represents the value
##### **Syntax :** plt.bar(
    x,
    height,
    width=0.8,
    color=None,
    edgecolor=None,
    linewidth=None,
    alpha=None,
    label=None,
    tick_label=None,
    hatch=None,
    bottom=None,
    align='center'
)
| Parameter  | Description      | Default Value |
| ---------- | ---------------- | ------------- |
| x          | Categories       | Required      |
| height     | Bar heights      | Required      |
| width      | Width of bars    | 0.8           |
| color      | Fill color       | None          |
| edgecolor  | Border color     | None          |
| linewidth  | Border thickness | None          |
| alpha      | Transparency     | None          |
| label      | Legend name      | None          |
| tick_label | Custom labels    | None          |
| hatch      | Pattern on bars  | None          |
| bottom     | Starting point   | None          |
| align      | Alignment        | 'center'      |

```python
import matplotlib.pyplot as plt
city = ['Banglore','Dharwad','Belagavi','Mysore','Hasan','Vijaypur','Chikkballapur','Manglore','Chikkmanglore','Shivmogga']
revenue = [115,76,68,85,78,65,34,82,98,93]
plt.bar(city,revenue, width=0.4, color='yellow',edgecolor='blue',linewidth=1,alpha=0.8,label='Revenue',hatch='*',align='center')
plt.title('City wise Revenue',fontsize=20,color='red',fontweight='bold')
plt.xlabel('City',fontsize=12,color='purple',fontweight='bold')
plt.ylabel('Revenue',fontsize=12,color='purple',fontweight='bold')
plt.xticks(rotation=90,color='blue')
plt.yticks(rotation=0,color='blue')
plt.grid(True,alpha=0.2)
plt.legend()
plt.show() 
```

#### **2. Horizontal Bar Chart (plt.barh()) :**
##### A Horizontal Bar Chart displays bars horizontally instead of vertically.
##### Categories are shown on the Y-axis
##### Values are shown on the X-axis
##### Length of the bar represents the value
##### Why Use Horizontal Bar Charts?
##### Horizontal bar charts are useful when:
##### Category names are long
##### There are many categories
##### Easy comparison of rankings
##### Better readability than vertical bars
##### **Data Analytics Use Cases**
| Category    | Value   |
| ----------- | ------- |
| Employees   | Salary  |
| Products    | Sales   |
| Cities      | Revenue |
| Departments | Budget  |
| Companies   | Profit  |
| Students    | Marks   |
##### **Syntax :** plt.barh(
    y,
    width,
    height=0.8,
    left=None,
    color=None,
    edgecolor=None,
    linewidth=None,
    alpha=None,
    label=None,
    hatch=None,
    align='center'
)
| Parameter | Description       | Default  |
| --------- | ----------------- | -------- |
| y         | Categories        | Required |
| width     | Values            | Required |
| height    | Bar thickness     | 0.8      |
| left      | Starting position | None     |
| color     | Fill color        | None     |
| edgecolor | Border color      | None     |
| linewidth | Border thickness  | None     |
| alpha     | Transparency      | None     |
| label     | Legend name       | None     |
| hatch     | Pattern           | None     |
| align     | Alignment         | center   |


```python
import matplotlib.pyplot as plt
city = ['Banglore','Dharwad','Belagavi','Mysore','Hasan','Vijaypur','Chikkballapur','Manglore','Chikkmanglore','Shivmogga']
revenue = [115,76,68,85,78,65,34,82,98,93]
plt.barh(city,revenue,height=0.5, color='yellow',edgecolor='blue',linewidth=1,alpha=0.8,label='Revenue',hatch='*',align='center',left=100)
plt.title('City wise Revenue',fontsize=20,color='red',fontweight='bold')
plt.xlabel('City',fontsize=12,color='purple',fontweight='bold')
plt.ylabel('Revenue',fontsize=12,color='purple',fontweight='bold')
plt.xticks(rotation=90,color='blue')
plt.yticks(rotation=0,color='blue')
plt.grid(True,alpha=0.2)
plt.legend()
plt.show() 
```

#### **4. Grouped Bar Chart (Clustered Bar Chart) :**
##### A Grouped Bar Chart displays multiple bars side-by-side for each category.
##### It is used to compare two or more datasets within the same category.
##### **Example :**
##### Compare Male and Female employees in each department.
| Department | Male | Female |
| ---------- | ---- | ------ |
| IT         | 30   | 25     |
| HR         | 20   | 15     |
| Sales      | 40   | 35     |
##### Instead of overlapping, bars are placed side-by-side.
##### **Why Use Grouped Bar Charts?**
##### Compare multiple datasets
##### Easy comparison
##### No overlapping
##### Most commonly used in Data Analytics
##### Dashboard friendly
| Category    | Dataset 1      | Dataset 2        |
| ----------- | -------------- | ---------------- |
| Departments | Male Employees | Female Employees |
| Months      | Sales 2024     | Sales 2025       |
| Products    | Online Sales   | Offline Sales    |
| Cities      | Revenue 2024   | Revenue 2025     |
| Students    | Mid Exam       | Final Exam       |
##### **Basic Syntax :** plt.bar(x1, y1, width)
##### plt.bar(x2, y2, width)
##### **Standard Syntax :**
##### import numpy as np
##### x = np.arange(len(categories))
##### width = 0.4
##### plt.bar(x-width/2, y1, width)
##### plt.bar(x+width/2, y2, width)
##### Understanding the Formula
##### Suppose: x = [0,1,2]
##### Then: x-width/2
##### moves bars left.
##### x+width/2
##### moves bars right.
##### Result:
##### | Bar1 | Bar2 |
##### Side-by-side comparison.

##### **Basic Syntax :**
##### plt.bar(x1, y1, width)
##### plt.bar(x2, y2, width

```python
import matplotlib.pyplot as plt
city = ['Banglore','Dharwad','Belagavi','Mysore','Hasan','Vijaypur','Chikkballapur','Manglore','Chikkmanglore','Shivmogga']
sales = [135,86,78,85,98,95,64,72,78,103]
profit = [55,76,48,78,78,77,34,70,68,93]
plt.bar(city,sales,width=0.5, color='blue',edgecolor='red',linewidth=1,alpha=0.7,label='sales')
plt.bar(city,profit,width=0.5, color='green',edgecolor='red',linewidth=1,alpha=0.7,label='profit')
plt.title('City wise Revenue',fontsize=20,color='red',fontweight='bold')
plt.xlabel('City',fontsize=12,color='purple',fontweight='bold')
plt.ylabel('Revenue',fontsize=12,color='purple',fontweight='bold')
plt.xticks(rotation=90,color='blue')
plt.yticks(rotation=0,color='blue')
plt.grid(True,alpha=0.2)
plt.legend()
plt.show() 
```

##### **Standard Syntax :**
##### import numpy as np
##### city=[...]
##### sales=[..]
##### profit=[...]
##### x = np.arange(len(categories))
##### width = 0.4
##### plt.bar(x-width/2, y1, width)
##### plt.bar(x+width/2, y2, width)
##### plt.xticks(x,city)

```python
import matplotlib.pyplot as plt
import numpy as np
city = ['Banglore','Dharwad','Belagavi','Mysore','Hasan','Vijaypur','Chikkballapur','Manglore','Chikkmanglore','Shivmogga']
sales = [135,86,78,85,98,95,64,72,78,103]
profit = [55,76,48,78,78,77,34,70,68,93]
x = np.arange(len(city)) #10
width = 0.4
plt.bar(x-width/2,sales,width, color='blue',edgecolor='k',linewidth=1,alpha=0.5,label='sales')  #10-0.4/2 = 9.8
plt.bar(x+width/2,profit,width, color='red',edgecolor='k',linewidth=1,alpha=0.5,label='profit') #10+0.4/2 = 10.2
plt.title('City wise Revenue',fontsize=20,color='red',fontweight='bold')
plt.xlabel('City',fontsize=12,color='purple',fontweight='bold')
plt.ylabel('Revenue',fontsize=12,color='purple',fontweight='bold')
plt.xticks(x,city,rotation=90,color='blue')
plt.yticks(rotation=0,color='blue')
plt.grid(True,alpha=0.2)
plt.legend()
plt.show() 
```

#### **5. Stacked Bar Chart :**
##### A Stacked Bar Chart displays multiple datasets stacked on top of each other in a single bar.
##### Instead of placing bars side-by-side (Grouped Bar Chart), the values are added vertically.
##### **Formula :** Total Height = Value1 + Value2 + Value3 + ...
##### **Why Use Stacked Bar Chart?**
##### Show total value
##### Show contribution of each category
##### Compare composition of data
##### Easy part-to-whole analysis
| Category  | Dataset 1 | Dataset 2 |
| --------- | --------- | --------- |
| Sales     | Online    | Offline   |
| Employees | Male      | Female    |
| Revenue   | Product A | Product B |
| Students  | Boys      | Girls     |
| Expenses  | Fixed     | Variable  |
##### **Basic Syntax :**
##### plt.bar(x, y1)
##### plt.bar(x, y2, bottom=y1)
##### **Complete Syntax :**
##### plt.bar(
    x,
    y1,
    width=0.8,
    color='blue',
    label='Dataset1'
)
##### plt.bar(
    x,
    y2,
    width=0.8,
    bottom=y1,
    color='green',
    label='Dataset2'
)

```python
import matplotlib.pyplot as plt
import numpy as np
city = ['Banglore','Dharwad','Belagavi','Mysore','Hasan','Vijaypur','Chikkballapur','Manglore','Chikkmanglore','Shivmogga']
online_class = [135,86,78,85,98,95,64,72,78,103]
offline_class = [155,176,148,178,178,177,134,170,168,193]
plt.bar(city,online_class,width=0.8, color='blue',edgecolor='k',linewidth=1,alpha=0.5,label='Online Class')  
plt.bar(city,offline_class,bottom=online_class,width=0.8, color='red',edgecolor='k',linewidth=1,alpha=0.5,label='Offline Class') 
plt.title('City wise Mode of learning',fontsize=20,color='red',fontweight='bold')
plt.xlabel('City',fontsize=12,color='purple',fontweight='bold')
plt.ylabel('Online & Offline Students Count',fontsize=12,color='purple',fontweight='bold')
plt.xticks(rotation=90,color='blue')
plt.yticks(rotation=0,color='blue')
plt.grid(True,alpha=0.2)
plt.legend()
plt.show() 
```
