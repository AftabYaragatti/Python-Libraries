# **Matplotlib Library for Data Analytics**

### **Violinplot in Matplotlib**
##### A Violin Plot combines the features of:
##### Box Plot
##### Histogram
##### Density Plot (KDE)
##### **It shows:** Distribution of data, Density of data, Median, Quartiles, Spread of data.
##### A Violin Plot gives more information than a Box Plot.
##### **What is a Violin Plot?**
##### A Violin Plot is called a "Violin" because its shape resembles a violin.
##### The width represents:
##### More Width --> More Data Points
##### Less Width --> Fewer Data Points
##### **Why Use Violin Plot?**
##### **A Box Plot only shows:** Min, Q1, Median, Q3, Max
##### A Violin Plot also shows: Density Distribution, So you can see where data is concentrated.
##### **Syntax :** plt.violinplot(
    dataset,
    positions=None,
    vert=True,
    widths=0.5,
    showmeans=False,
    showmedians=False,
    showextrema=True,
    points=100
)

```python
import matplotlib.pyplot as plt
marks=[94,87,89,83,54,76,64,96,45,32,94,85]
plt.violinplot(marks)
plt.show()
```

```python
import matplotlib.pyplot as plt
sales_2024=[84,45,55,51,72,62,81,69,92,37,53,99]
plt.violinplot(sales_2024)
plt.show()
```

| Violin Plot                      | Box Plot               |
