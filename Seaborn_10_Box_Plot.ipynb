# **Seaborn Library for Data Analytics**

#### **What is a Box Plot :**
##### A Box Plot (Box-and-Whisker Plot) is used to visualize the distribution of numerical data and identify:
##### Minimum Value
##### Maximum Value
##### Median (Q2)
##### First Quartile (Q1)
##### Third Quartile (Q3)
##### Outliers
##### **Definition :** A Box Plot summarizes the distribution of a dataset using quartiles and highlights outliers.
##### 1. Minimum
##### Smallest value (excluding outliers)
##### 2. Maximum
##### Largest value (excluding outliers)
##### 3. Median (Q2)
##### Middle value of dataset
##### 4. First Quartile (Q1)
##### 25% of data lies below Q1
##### 5. Third Quartile (Q3)
##### 75% of data lies below Q3
##### 6. IQR (Inter Quartile Range)
##### IQR = Q3 - Q1
#### **Why Use Box Plot :**
##### Detect Outliers
##### Compare Distributions
##### Check Data Spread
##### Identify Skewness
##### Statistical Analysis
##### Machine Learning EDA
##### Data Cleaning

##### **Syntax :** sns.boxplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    order=None,
    hue_order=None,
    orient=None,
    color=None,
    palette=None,
    saturation=0.75,
    width=0.8,
    dodge='auto',
    whis=1.5,
    linewidth=None,
    fill=True,
    legend='auto'
)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
df=pd.read_csv('Employee_Dataset.csv')
df
```

##### **x :** Categorical variable.
##### **y :** Numerical variable.

```python
sns.boxplot(data=df,x='Department',y='Experience_Years')
plt.title('Box Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Years of Experiance',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.2)
plt.tight_layout()
plt.show()
```

#### **whis :**
##### Controls whisker length.
##### **Formula :**
##### Lower Limit = Q1 - 1.5 × IQR
##### Upper Limit = Q3 + 1.5 × IQR
##### Values outside this range become outliers.

```python
sns.boxplot(data=df,x='Department',y='Experience_Years',hue='Gender',whis=2)
plt.title('Box Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Years of Experiance',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.2)
plt.tight_layout()
plt.show()
```
