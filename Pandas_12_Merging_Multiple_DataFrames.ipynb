# **Pandas Library**
### **Merging Multiple DataFrames**

##### **Merging Multiple DataFrames :**
##### To merge multiple DataFrames in pandas we have mainly 3 different ways.
##### 1.Concat
##### 2.Merge
##### 3.Join

```python
import pandas as pd
import numpy as np
df1=pd.read_csv("Employee_Cleaned_Dataset.csv")
df1.drop(columns='Unnamed: 0', inplace=True)
df1
```

```python
df2=pd.read_csv("Employee_Cleaned_Dataset.csv")
df2.drop(columns='Unnamed: 0', inplace=True)
df2
```

##### **1. Concat() :**
##### It is used to merge multiple DataFrames in both ways (row_wise,col_wise)
##### To perform concatenation we use pandas inbuilt method **concat()**
##### Concatenation has been classified into two types :
##### 1.Vertical Concatenation() - Row Wise
##### 2.Horizontal Concatenation() - Column Wise

##### **1.1 Vertical Concatenation() - Row Wise :**
##### While performing vertical concatenation, it is manadatory to check whether we have same column names in all the dataframes. If we have different column names it returns null values.
##### **Syntax : pd.concat([DataFrame1,DataFrame2....],axis=0,ignore_index=True)**

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],axis=0)
```

```python
pd.concat([df1,df2],axis=0,ignore_index=True)
```

```python
pd.concat([df1,df2],axis=0,ignore_index=False)
```

##### **1.2 Horizontal Concatenation() - Column Wise :**
##### While performing Horizontal concatenation,If we have any common column names it will create duplicates of those columns.If we have totally different column names then we go with column wise concatenation.
##### **Syntax : pd.concat([DataFrame1,DataFrame2....],axis=1,ignore_index=True)**

```python
pd.concat([df1,df2],axis=1)
```

```python
pd.concat([df1,df2],axis=1,ignore_index=True)
```

##### **NOTE :**
##### When we use ignore_index=True, It returns column index instead of column names.
##### When we have duplicate columns we use :
##### a) df.T.drop_duplicates().T # Transpose
##### b) df.loc[:,~(df.columns.duplicated())] # Bitwise
##### c) Selecting columns(using fancy indexing) :
##### df=df[['col1','col2','col3']] or df=df.iloc[:,[col_index,col_index,....]]

##### **2. Merge :**
##### It is used to merge two dataframes. Merge will perform internally SQL style join operations.
##### **Syntax : pd.merge(df1,df2,on='common_column_name',how='inner/outer/left/right/cross')**
##### The default value of how is inner, If we are not providing any common column name it takes by default first common column name.

```python
df1 = pd.DataFrame({
    'ID': [101, 102, 103],
    'Name': ['Aftab', 'John', 'Alice']
})
df1
```

```python
df2 = pd.DataFrame({
    'ID': [102, 103, 104],
    'Salary': [50000, 60000, 70000]
})
df2
```

```python
pd.merge(df1, df2, on='ID')
```

##### **Inner Join (Default)** : Returns only matching records from both DataFrames.

```python
pd.merge(df1, df2, on='ID', how='inner')
```

##### **Outer Join :** Returns all records from both DataFrames.

```python
pd.merge(df1, df2, on='ID', how='outer')
```

##### **Left Join :** Returns all records from the left DataFrame (df1) and matching records from df2

```python
pd.merge(df1, df2, on='ID', how='left')
```

##### **Right Join :** Returns all records from the right DataFrame (df2) and matching records from df1.

```python
pd.merge(df1, df2, on='ID', how='right')
```

##### **Cross Join :** Creates every possible combination of rows from both DataFrames.

```python
pd.merge(df1, df2, how='cross')
```

##### **NOTE :**
##### inner → Matching rows only
##### left → All rows from left DataFrame
##### right → All rows from right DataFrame
##### outer → All rows from both DataFrames
##### cross → Cartesian product (all possible combinations)

##### **3. Join() :**
##### It is used to join dataframes based on index. Join will perform index style operation. For join internally performs left join operations.
##### **Syntax : df1.join(df2,how='inner/outer/left/right')**

```python
df1 = pd.DataFrame({
    'EID': [101, 102, 103],
    'Name': ['Aftab','Priya','Alice']
})
df1
```

```python
df2 = pd.DataFrame({
    'ID': [102, 103, 104],
    'Salary': [50000, 60000, 70000]
})
df2
```

```python
df1.join(df2)
```

```python
df1.join(df2,how='left')
```

```python
df1.join(df2,how='right')
```

```python
df1.join(df2,how='inner')
```

```python
df1.join(df2,how='outer')
```

```python
df1.join(df2,how='cross')
```

##### **Note :**
##### join() joins DataFrames using their index by default.
##### It is similar to merge(), but simpler when joining on indexes.
##### Default value of how is 'left'.
