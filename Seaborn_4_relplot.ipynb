# **Seaborn Library for Data Analytics**

#### **What is relplot() :**
##### relplot() stands for Relational Plot. It is a figure-level function in Seaborn used to visualize relationships between variables.
##### **Important :** relplot() internally uses: scatterplot() & lineplot(), Depending on the kind parameter.
##### relplot() is a high-level Seaborn function used to create scatter plots or line plots and easily visualize relationships among multiple variables.
##### **Why Use relplot() :**
##### Create Scatter Plot and Line Plot using one function
##### Easy grouping using hue
##### Easy faceting using row and col
##### Better for large datasets
##### Creates multiple plots automatically
##### Most useful in EDA (Exploratory Data Analysis)
##### Default: kind='scatter'
##### So it creates a Scatter Plot.

#####  **Syntax :** sns.relplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    size=None,
    style=None,
    units=None,
    weights=None,
    row=None,
    col=None,
    col_wrap=None,
    row_order=None,
    col_order=None,
    palette=None,
    hue_order=None,
    sizes=None,
    size_order=None,
    markers=True,
    dashes=True,
    legend='auto',
    kind='scatter',
    height=5,
    aspect=1
)

#### **dashes :** Used only when kind=line
##### Controls dashed lines.(True/False)
##### Seaborn uses different line styles:
##### Solid line (-)
##### Dashed line (--)
##### Dotted line (:)
##### Dash-dot line (-.)
##### dashes=True + linestyle=':'
##### These can conflict.
##### When you use: dashes=True
##### Seaborn automatically assigns different dash patterns.
##### Then: linestyle=':' tries to make all lines dotted.So choose one.

```python
sns.relplot(data=df, x='Degree Yop', y='Degree Percentage',hue='Courses Taken',size='Admission Type',palette='deep',sizes=(2,5),marker='p',markers=True,kind='line',linestyle=':')
plt.title('Relplot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.xlabel('Degree YOP',color='purple',fontsize=12)
plt.ylabel('Degree Percentage',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True)
plt.legend(loc='best',bbox_to_anchor=(1.05,1),)
plt.tight_layout()
plt.show()
```

#### **legend :**
##### Controls legend display.
##### **Values**
##### legend='auto'
##### legend='brief'
##### legend='full'
##### legend=False
##### **legend='auto' (Default)**
##### Seaborn automatically decides how much information to show.
##### For a small number of categories → shows all categories.
##### For many categories → may simplify the legend.
##### **legend='brief'**
##### Shows a summarized legend.
##### Mainly useful when the hue variable is numeric and contains many unique values.
##### **legend='full'**
##### Displays every unique value in the legend.
##### legend=False : Removes the legend completely.
| Value | Meaning                           | Best For           |
| ----- | --------------------------------- | ------------------ |
| auto  | Seaborn decides                   | General use        |
| brief | Shows fewer representative values | Large numeric data |
| full  | Shows all values                  | Detailed analysis  |
| False | No legend                         | Clean plots        |

```python
sns.relplot(data=df, x='Degree Yop', y='Degree Percentage',hue='Courses Taken',size='Admission Type',legend='full')
plt.title('Relplot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.xlabel('Degree YOP',color='purple',fontsize=12)
plt.ylabel('Degree Percentage',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True)
plt.legend(loc='best',bbox_to_anchor=(1.05,1),)
plt.tight_layout()
plt.show()
```

#### **col_wrap :**
##### Wrap columns into multiple rows.
##### **Example :** col_wrap=2
##### **Output :**
##### Instead of
##### A B C D
##### Produces
##### A B
##### C D

```python
g=sns.relplot(data=df, x='Degree Yop', y='Degree Percentage',hue='Admission Type',col='Courses Taken',row_order=['Python Selenium','Python Full Stack','Python Data Analytics','Python Data Science'],col_wrap=2)
g.set_axis_labels('Degree YOP','Degree Percentage')
g.fig.suptitle('Relplot',fontsize=15,color='red')
for ax in g.axes.flat:
    ax.set_xlabel('Degree Yop',color='purple',fontsize=12)
    ax.set_ylabel('Degree Percentage',color='purple',fontsize=12)
    ax.tick_params(axis='x',color='blue',rotation=90)
    ax.tick_params(axis='y',color='blue',rotation=0)
    ax.grid(True)
g.fig.tight_layout(rect=[0, 0, 1, 0.96])
```
