# **Seaborn Library for Data Analytics**

#### **What is a Violin Plot :**
##### A Violin Plot combines features of:
##### Box Plot
##### KDE Plot (Kernel Density Estimation)
##### It shows both:
##### Statistical summary (Median, Quartiles)
##### Data distribution (Density)
##### **Definition :** A Violin Plot visualizes the distribution of numerical data and its probability density.
#### **Why is it called a Violin Plot :**
##### Because its shape resembles a violin musical instrument.
      /\
     /  \
    /    \
   |      |
    \    /
     \  /
      \/
##### The wider the violin, the higher the concentration of data.
#### **Why Use Violin Plot :**
##### Distribution Analysis
##### Compare Multiple Categories
##### Density Visualization
##### Outlier Detection
##### Statistical Analysis
##### Machine Learning EDA
##### Alternative to Box Plot
#### **Violin Plot vs Box Plot :**
| Box Plot         | Violin Plot               |
| ---------------- | ------------------------- |
| Shows Quartiles  | Shows Quartiles + Density |
| Simple           | More Detailed             |
| Outliers Visible | Distribution Visible      |
| Less Information | More Information          |
##### **Example :** Suppose two departments have the same median salary.
##### A Box Plot shows:
##### IT      HR
##### |----|  |----|
##### A Violin Plot shows:
##### IT        HR
##### <====>   <==>
##### You can see how data is distributed.
##### **Syntax :** sns.violinplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    order=None,
    hue_order=None,
    orient=None,
    color=None,
    palette=None,
    saturation=0.75,
    inner='box',
    split=False,
    density_norm='area',
    common_norm=False,
    fill=True,
    linewidth=None,
    width=0.8,
    dodge='auto',
    cut=2,
    legend='auto'
)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
df=pd.read_csv('Employee_Dataset.csv')
df
```

##### **X :** Categorical Variable
##### **Y :** Numerical Variable

```python
sns.violinplot(data=df,x='Department',y='Experience_Years')
plt.title('Violin Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Experience',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
sns.violinplot(data=df,x='Gender',y='Performance_Rating')
plt.title('Violin Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Performance Rating',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.2)
plt.tight_layout()
plt.show()
```
