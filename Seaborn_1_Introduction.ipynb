# **Seaborn Library for Data Analytics**

### **Seaborn in Data Analytics**
##### **What is Seaborn :** Seaborn is a Python Data Visualization Library built on top of Matplotlib.
##### **It provides:**
##### Beautiful statistical graphics
##### Attractive default themes
##### Advanced visualization techniques
##### Easy integration with Pandas DataFrames
##### Less code compared to Matplotlib
##### **Definition :** Seaborn is a high-level Python library used for creating attractive, informative, and statistical visualizations.
##### **Import Syntax :**
##### import seaborn as sns
##### import matplotlib.pyplot as plt
#### **Why Seaborn Over Matplotlib?**
| Matplotlib                           | Seaborn                           |
| ------------------------------------ | --------------------------------- |
| Low-level library                    | High-level library                |
| More code required                   | Less code required                |
| Basic visual appearance              | Beautiful default appearance      |
| Manual styling needed                | Automatic styling                 |
| Good for simple plots                | Good for statistical plots        |
| Difficult for complex visualizations | Easy for complex visualizations   |
| Less integration with DataFrames     | Excellent integration with Pandas |

#### **Example : Matplotlib**

```python
import matplotlib.pyplot as plt
x=[1,4,6,8]
y=[10,15,30,40]
plt.plot(x,y)
plt.show()
```

#### **Example : Seaborn**

```python
import seaborn as sns
import pandas as pd
df=pd.DataFrame({
    'x':[1,5,6,8],
    'y':[10,15,35,40]
})
sns.lineplot(data=df,x='x',y='y')
plt.show()
```

##### **9. Box Plot :** sns.boxplot()

##### **13. Point Plot :** sns.pointplot()

##### **14. Boxen Plot :** sns.boxenplot()

##### **D. Matrix Plots :** Used for correlation analysis.
##### **15. Heatmap :** sns.heatmap()

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Creating Dataset
df = pd.DataFrame({
    "Name": ["Aftab", "Rahul", "Priya", "Sneha", "Aman",
             "Riya", "Kiran", "Pooja", "Arjun", "Neha"],
    
    "Age": [21, 22, 20, 23, 24,
            22, 21, 25, 23, 24],
    
    "Salary": [25000, 30000, 28000, 40000, 45000,
               35000, 32000, 50000, 42000, 47000],
    
    "Gender": ["Male", "Male", "Female", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],
    
    "Experience": [1, 2, 1, 3, 4,
                   2, 2, 5, 3, 4],
    
    "Department": ["IT", "HR", "IT", "Sales", "IT",
                   "HR", "Sales", "IT", "HR", "Sales"]
})

# Theme
sns.set_theme(style="whitegrid")

# Figure Size
plt.figure(figsize=(10,6))

# Scatter Plot
sns.scatterplot(
    data=df,                 # Dataset
    x="Age",                 # X-axis
    y="Salary",              # Y-axis
    hue="Gender",            # Color grouping
    style="Department",      # Marker style
    size="Experience",       # Marker size
    palette="Set2",          # Color theme
    sizes=(100,500),         # Min-Max marker size
    alpha=0.8               # Transparency
)

# Title
plt.title("Employee Salary Analysis", fontsize=16)

# Axis Labels
plt.xlabel("Age", fontsize=12)
plt.ylabel("Salary", fontsize=12)

# Grid
plt.grid(True)

# Legend
plt.legend(title="Categories")

# Show Plot
plt.show()
```
