# **Matplotlib Library for Data Analytics**

### **Pie Charts in Matplotlib :**
##### A Pie Chart is a circular chart divided into slices (wedges), where each slice represents a proportion or percentage of the whole.
##### **Formula :** Percentage = (Category Value / Total Value) × 100
##### **Example :**
| Product | Sales |
| ------- | ----- |
| Laptop  | 40    |
| Mobile  | 30    |
| TV      | 20    |
| Tablet  | 10    |
##### Total = 100
##### Laptop = 40%
##### Mobile = 30%
##### TV = 20%
##### Tablet = 10%
##### **Uses of Pie Charts :**
| Use Case                | Example                   |
| ----------------------- | ------------------------- |
| Market Share            | Samsung vs Apple          |
| Sales Distribution      | Product-wise Sales        |
| Budget Allocation       | Department-wise Budget    |
| Employee Distribution   | Department-wise Employees |
| Revenue Contribution    | Product-wise Revenue      |
| Population Distribution | City-wise Population      |
##### **Best Used When :**
##### Showing percentages
##### Showing part-to-whole relationships
##### Number of categories ≤ 5 or 6
##### **Syntax :** plt.pie(
    x,
    explode=None,
    labels=None,
    colors=None,
    autopct=None,
    pctdistance=0.6,
    shadow=False,
    labeldistance=1.1,
    startangle=0,
    radius=1,
    counterclock=True,
    wedgeprops=None,
    textprops=None,
    center=(0,0),
    frame=False,
    rotatelabels=False
)
| Parameter     | Description         | Default Value  |
| ------------- | ------------------- | -------------- |
| x             | Data values         | Required       |
| labels        | Slice labels        | None           |
| colors        | Slice colors        | Default colors |
| autopct       | Show percentage     | None           |
| explode       | Separate slice      | None           |
| shadow        | Add shadow          | False          |
| startangle    | Starting angle      | 0              |
| radius        | Pie size            | 1              |
| pctdistance   | Percentage position | 0.6            |
| labeldistance | Label position      | 1.1            |
| counterclock  | Direction           | True           |
| wedgeprops    | Slice properties    | None           |
| textprops     | Text properties     | None           |
| center        | Pie center          | (0,0)          |
| frame         | Frame display       | False          |
| rotatelabels  | Rotate labels       | False          |


#### **labels Parameter :**

```python
import matplotlib.pyplot as plt
dept = ['IT','Sales','Business','Financial','Delivery']
no_of_emp = [134,150,45,60,180]
plt.pie(no_of_emp, labels=dept)
plt.title('Number of Wmployees in Each Department',fontsize=12,color='red',fontweight='bold')
plt.legend(loc='best',bbox_to_anchor=(1,1))
plt.show() 
```

#### **explode :**
##### The explode parameter is used to separate one or more slices from the center of the pie chart.
##### It is commonly used to highlight an important category.
##### **Syntax :** plt.pie(x, explode=explode_values)
| Value | Meaning               |
| ----- | --------------------- |
| 0     | No separation         |
| 0.1   | Slight separation     |
| 0.2   | Medium separation     |
| 0.3   | Large separation      |
| 0.5   | Very large separation |
##### **Example :** explode = [0, 0, 0, 0]

```python
import matplotlib.pyplot as plt
dept = ['IT','Sales','Business','Financial','Delivery']
no_of_emp = [134,150,45,60,180]
plt.pie(no_of_emp, labels=dept, explode=[0,0,0,0,0.1])
plt.title('Number of Wmployees in Each Department',fontsize=12,color='red',fontweight='bold')
plt.legend(loc='best',bbox_to_anchor=(1,1))
plt.show() 
```

#### **colors :**
##### The colors parameter is used to assign custom colors to pie chart slices.
##### By default, Matplotlib automatically chooses colors. Using colors, you can control exactly which color each slice uses.
##### **Syntax :** plt.pie(x,colors=color_list)

```python
import matplotlib.pyplot as plt
dept = ['IT','Sales','Business','Financial','Delivery']
no_of_emp = [134,150,45,60,180]
plt.pie(no_of_emp, labels=dept, explode=[0,0,0,0,0.1],colors=['yellow','green','cyan','blue','purple'])
plt.title('Number of Wmployees in Each Department',fontsize=12,color='red',fontweight='bold')
plt.legend(loc='best',bbox_to_anchor=(1,1))
plt.show() 
```

#### **autopct :**
##### The autopct parameter automatically displays the percentage value of each slice on the pie chart.
##### Without autopct, only the slices are shown.
##### With autopct, percentages like 25.0%, 40.5%, etc., are displayed.
##### **Syntax :** plt.pie(x, autopct='format')
| Symbol | Meaning               |
| ------ | --------------------- |
| `%`    | Start format          |
| `m`    | Minimum width         |
| `n`    | Decimal places        |
| `f`    | Floating point number |
| `%%`   | Percent symbol (%)    |
##### **Example :** autopct='%1.1f%%'

```python
import matplotlib.pyplot as plt
dept = ['IT','Sales','Business','Financial','Delivery']
no_of_emp = [134,150,45,60,180]
plt.pie(no_of_emp, labels=dept,autopct='%1.2f%%')
plt.title('Number of Wmployees in Each Department',fontsize=12,color='red',fontweight='bold')
plt.legend(loc='best',bbox_to_anchor=(1,1))
plt.show() 
```

#### **pctdistance :**
##### The pctdistance parameter controls the position of the percentage text (autopct) inside or outside the pie chart slices.
##### It specifies the distance of the percentage text from the center of the pie.
##### **Syntax :** plt.pie(x, autopct='%1.1f%%', pctdistance=value)
Center (0)
    |
    |
0.3 |---- Percentage near center
    |
0.6 |---- Default position
    |
0.8 |---- Near slice edge
    |
1.0 |---- On slice boundary

```python
import matplotlib.pyplot as plt
dept = ['IT','Sales','Business','Financial','Delivery']
no_of_emp = [134,150,45,60,180]
plt.pie(no_of_emp, labels=dept,autopct='%1.2f%%',pctdistance=0.8)
plt.title('Number of Wmployees in Each Department',fontsize=12,color='red',fontweight='bold')
plt.legend(loc='best',bbox_to_anchor=(1,1))
plt.show() 
```

#### **shadow :**

#### **labeldistance :**

##### It determines where the labels (e.g., Bangalore, Mysore, Laptop, Mobile) are placed relative to the pie slices.

#### **startangle :**
##### The startangle parameter is used to rotate the entire pie chart.
##### By default, the first slice starts at 0° (3 o'clock position).
##### Using startangle, you can change where the first slice begins.
##### **Understanding angels :**
          90°
           ↑
           |
180° ←     ○     → 0°
           |
           ↓
          270°

```python
import matplotlib.pyplot as plt
dept = ['IT','Sales','Business','Financial','Delivery']
no_of_emp = [134,150,45,60,180]
plt.pie(no_of_emp, labels=dept,autopct='%1.2f%%',pctdistance=0.6,labeldistance=1.1,startangle=90)
plt.title('Number of Wmployees in Each Department',fontsize=12,color='red',fontweight='bold')
plt.legend(loc='best',bbox_to_anchor=(1,1))
plt.show() 
```

#### **wedgeprops :**
##### The wedgeprops parameter is used to customize the appearance of pie chart slices (called wedges).
##### Using wedgeprops, you can control:
##### Border color
##### Border width
##### Border style
##### Slice width (Donut Chart)
##### Transparency
##### Fill color behavior
##### **Syntax :** plt.pie(x, wedgeprops={})
| Property  | Description           | Default |
| --------- | --------------------- | ------- |
| edgecolor | Border color          | None    |
| linewidth | Border thickness      | 1       |
| linestyle | Border style          | '-'     |
| width     | Donut chart hole size | None    |
| alpha     | Transparency          | 1       |
| fill      | Fill slice or not     | True    |

```python
import matplotlib.pyplot as plt
dept = ['IT','Sales','Business','Financial','Delivery']
no_of_emp = [134,150,45,60,180]
plt.pie(no_of_emp, labels=dept,autopct='%1.2f%%',pctdistance=0.6,labeldistance=1.1,startangle=90,wedgeprops={'edgecolor':'black'})
plt.title('Number of Wmployees in Each Department',fontsize=12,color='red',fontweight='bold')
plt.legend(loc='best',bbox_to_anchor=(1,1))
plt.show() 
```

#### **center :**
##### The center parameter is used to change the position of the center of the pie chart.
##### By default, the pie chart is drawn at the center of the axes.
##### Using center, you can move the pie chart: Left  Right  Up Down
##### **Syntax :** plt.pie(x, center=(x_coordinate, y_coordinate))
##### center=(x,y) : Default : (0,0)
| Parameter | Meaning             |
| --------- | ------------------- |
| x         | Horizontal Position |
| y         | Vertical Position   |
| Value        | Result           |
| ------------ | ---------------- |
| `(0,0)`      | Center (Default) |
| `(1,0)`      | Right            |
| `(-1,0)`     | Left             |
| `(0,1)`      | Up               |
| `(0,-1)`     | Down             |
| `(0.5,0.5)`  | Upper Right      |
| `(-0.5,0.5)` | Upper Left       |

```python
import matplotlib.pyplot as plt
dept = ['IT','Sales','Business','Financial','Delivery']
no_of_emp = [134,150,45,60,180]
plt.pie(no_of_emp, labels=dept,autopct='%1.2f%%',pctdistance=0.6,labeldistance=1.1,startangle=90,wedgeprops={'edgecolor':'black','linewidth':1,'linestyle':'--','width':0.8,'alpha':0.7,'fill':True},
       textprops={'fontsize':12, 'color':'blue', 'fontweight':'bold', 'fontstyle':'italic', 'alpha':1, 'fontfamily':'Times New Roman', 'rotation':30},center=(1,0))
plt.title('Number of Wmployees in Each Department',fontsize=12,color='red',fontweight='bold')
plt.legend(loc='best',bbox_to_anchor=(0,0))
plt.show() 
```

```python
import matplotlib.pyplot as plt

products = ['Laptop','Mobile','TV','Tablet']
sales = [40,30,20,10]

plt.pie(
    sales,
    labels=products,
    autopct='%1.1f%%',
    startangle=180,
    colors=['gold','skyblue','lightgreen','orange']
)

plt.axis('equal')

plt.title(
    'Product Sales Distribution',
    fontsize=14,
    color='red',
    fontweight='bold'
)

plt.show()
```

#### **3D Style Pie Chart (Visual Effect) :**

```python
import matplotlib.pyplot as plt
dept = ['IT','Sales','Business','Financial','Delivery']
no_of_emp = [134,150,45,60,180]
plt.pie(no_of_emp, labels=dept, explode=[0.1,0,0,0,0],shadow=True, colors=['gold','skyblue','lightgreen','orange','pink'],autopct='%1.2f%%',pctdistance=0.6,labeldistance=1.2,startangle=90,
       textprops={'fontsize':12, 'color':'blue', 'fontweight':'bold', 'fontstyle':'italic', 'alpha':1, 'fontfamily':'Times New Roman', 'rotation':30},center=(0,0))
plt.title('Number of Wmployees in Each Department',fontsize=12,color='red',fontweight='bold')
plt.legend(loc='best',bbox_to_anchor=(0,0))
plt.show() 
```
