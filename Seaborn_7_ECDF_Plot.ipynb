#### **What is ECDF Plot :**
##### ECDF stands for Empirical Cumulative Distribution Function.
##### An ECDF Plot shows the cumulative proportion (or percentage) of observations less than or equal to a given value.
##### **Definition :** ECDF is a graph that displays the cumulative distribution of data points.

#### **Example :** 
##### Salary Data = [10000, 20000, 30000, 40000, 50000]
| Salary | Cumulative Percentage |
| ------ | --------------------- |
| 10000  | 20%                   |
| 20000  | 40%                   |
| 30000  | 60%                   |
| 40000  | 80%                   |
| 50000  | 100%                  |

#### **Why Use ECDF Plot :**
##### Understand cumulative distribution
##### Find percentiles
##### Compare distributions
##### Detect skewness
##### Statistical analysis
##### Machine Learning EDA
##### Alternative to Histogram and KDE

##### **Syntax :** sns.ecdfplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    weights=None,
    stat='proportion',
    complementary=False,
    palette=None,
    color=None,
    linewidth=2,
    linestyle='-',
    alpha=None,
    legend=True
)

```python
import pandas as pd
import numpy as np

np.random.seed(42)

data = {
    'Emp_ID': range(1001, 1051),
    'Employee_Name': [
        'Aftab','Rahul','Priya','Sneha','Arjun','Kiran','Rohit','Anjali','Vikas','Pooja',
        'Deepak','Neha','Aman','Shweta','Nikhil','Kavya','Manoj','Riya','Suresh','Divya',
        'Harish','Megha','Akash','Nandini','Yash','Preeti','Varun','Isha','Rakesh','Komal',
        'Abhishek','Aarti','Sanjay','Monika','Tarun','Bhavna','Vivek','Simran','Ajay','Sonal',
        'Raj','Payal','Karthik','Pallavi','Girish','Anu','Sameer','Mansi','Nitin','Shruti'
    ],
    'Gender': np.random.choice(['Male','Female'], 50),
    'Age': np.random.randint(22, 45, 50),
    'Department': np.random.choice(
        ['Data Analytics','Data Science','HR','Finance','IT','Marketing'],
        50
    ),
    'Education': np.random.choice(
        ['BCA','BSc','BTech','MCA','MTech','MBA'],
        50
    ),
    'Experience_Years': np.random.randint(0, 15, 50),
    'Salary_LPA': np.round(np.random.uniform(3.5, 18, 50), 2),
    'Performance_Rating': np.random.randint(1, 6, 50),
    'Joining_Year': np.random.randint(2015, 2026, 50),
    'City': np.random.choice(
        ['Bangalore','Mumbai','Pune','Hyderabad','Chennai'],
        50
    ),
    'Work_Mode': np.random.choice(
        ['Work From Office','Hybrid','Remote'],
        50
    ),
    'Project_Count': np.random.randint(1, 12, 50),
    'Training_Hours': np.random.randint(10, 200, 50),
    'Promotion_Status': np.random.choice(
        ['Promoted','Not Promoted'],
        50
    ),
    'Attrition': np.random.choice(
        ['Yes','No'],
        50
    )
}
df = pd.DataFrame(data)
print(df.head())
```

##### **x** : Numerical variable displayed on X-axis.
##### **y** : Creates ECDF using Y-axis variable.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.ecdfplot(data=df,x='Experience_Years')
plt.title('ECDF Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Number of Counts',color='purple',fontsize=12)
plt.xlabel('Years of Experience',color='purple',fontsize=12)
plt.yticks(color='blue',rotation=90)
plt.xticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.2)
plt.tight_layout()
plt.show()
```

| ECDF                   | KDE                         |
| ---------------------- | --------------------------- |
| Cumulative probability | Probability density         |
| Always increasing      | Peaks and valleys           |
| Good for percentiles   | Good for distribution shape |

| ECDF                       | Histogram               |
| -------------------------- | ----------------------- |
| Cumulative distribution    | Frequency distribution  |
| Smooth increasing curve    | Bars                    |
| Easy percentile analysis   | Easy frequency analysis |
| Statistical interpretation | Distribution overview   |
