# **Seaborn Library for Data Analytics**

#### **What is a Bar Plot :**
##### A Bar Plot is used to compare numerical values across different categories.
##### It displays data using rectangular bars whose heights (or lengths) represent values.
##### **Definition :** A Bar Plot visualizes the relationship between a categorical variable and a numerical variable.
#### **Why Use Bar Plot :**
##### Compare categories
##### Compare averages
##### Department-wise analysis
##### Sales comparison
##### Revenue comparison
##### Business reporting
##### Dashboard visualization
##### Machine Learning EDA
##### **Syntax :** sns.barplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    order=None,
    hue_order=None,
    estimator='mean',
    errorbar=('ci',95),
    palette=None,
    color=None,
    saturation=0.75,
    width=0.8,
    dodge='auto',
    orient=None,
    alpha=None,
    legend='auto'
)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
df=pd.read_csv('Employee_Dataset.csv')
df
```

##### **X :** Categorical variable displayed on X-axis.
##### **Y :** Numerical variable displayed on Y-axis.

```python
sns.barplot(data=df,x='Department',y='Experience_Years')
plt.title('Bar Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Years of Experiance',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=0)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
sns.barplot(data=df,x='Experience_Years',y='Department')
plt.title('Bar Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.xlabel('Years of Experiance',color='purple',fontsize=12)
plt.ylabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=0)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.2)
plt.tight_layout()
plt.show()
```

#### **estimator :**
##### Determines how values are aggregated.
##### Default : estimator=np.mean
##### Values :** np.mean, np.median, np.sum, np.min, np.max

```python
import numpy as np
sns.barplot(data=df,x='Department',y='Experience_Years',hue='Gender',hue_order=['Female','Male'],estimator=np.sum)
plt.title('Bar Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Years of Experiance',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.2)
plt.tight_layout()
plt.show()
```

#### **errorbar :**
##### Displays uncertainty/error bars.
##### **Values :**
##### errorbar=('ci',95)
##### errorbar=('ci',99)
##### errorbar='sd'
##### errorbar=None

```python
import numpy as np
sns.barplot(data=df,x='Department',y='Experience_Years',hue='Gender',hue_order=['Female','Male'],errorbar=('ci',35))
plt.title('Bar Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Years of Experiance',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.2)
plt.tight_layout()
plt.show()
```
