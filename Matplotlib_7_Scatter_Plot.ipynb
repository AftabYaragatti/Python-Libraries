# **Matplotlib Library for Data Analytics**

### **Scatter Plot in Matplotlib**
##### As a Data Analyst, Scatter Plot is one of the most important charts because it helps us understand the relationship (correlation) between two numerical variables.
##### **Example:**
##### Hours Studied vs Marks
##### Age vs Salary
##### Advertising Cost vs Sales
##### Experience vs Income
##### **What is a Scatter Plot?** A Scatter Plot displays individual data points on a graph.
##### Each point has:
##### X Coordinate → X-axis value
##### Y Coordinate → Y-axis value
##### **Example :**
| Student | Hours Studied (X) | Marks (Y) |
| ------- | ----------------- | --------- |
| A       | 2                 | 35        |
| B       | 4                 | 50        |
| C       | 6                 | 65        |
| D       | 8                 | 80        |
##### **Why Use Scatter Plot?**
##### Scatter Plot is used to:
##### Find Correlation
##### Detect Outliers
##### Identify Trends
##### Compare Variables
##### Visualize Distribution
##### Machine Learning Data Analysis

##### **Syntax :** plt.scatter(
    x,
    y,
    s=None,
    c=None,
    marker=None,
    cmap=None,
    alpha=None,
    linewidths=None,
    edgecolors=None,
    label=None 
)

```python
import matplotlib.pyplot as plt
student = ['Aftab','Smith','Scott','Jhon','Alice','King','Ajay','Sam','Roy','Bob']
marks = [95,87,94,76,34,67,78,84,93,84]
hours_studied = [6,5,8,4,5,7,3,7,3,4]
plt.scatter(marks,hours_studied)
plt.title('Student Marks wrt study hours',color='red',fontsize=20)
plt.xlabel('Student Marks',color='blue',fontsize=12)
plt.ylabel('Student Studied in Hours',color='blue',fontsize=12)
plt.xticks(color='green')
plt.yticks(color='green')
plt.grid(True)
plt.show()
```

### **Grouped Scatter Plot :**

```python
import matplotlib.pyplot as plt
student = ['Aftab','Smith','Scott','Jhon','Alice','King','Ajay','Sam','Roy','Bob']
marks_male = [95,87,94,76,34,67,78,84,93,84]
marks_female = [98,57,84,73,64,87,85,82,99,67]
hours_studied_male = [6,5,8,4,5,7,3,7,3,4]
hours_studied_female = [9,4,3,6,8,8,5,3,7,2]
plt.scatter(marks_male,hours_studied_male,s=100,c='red',marker='*',linewidth=2,label='Male Students')
plt.scatter(marks_female,hours_studied_female,s=100,c='blue',marker='*',linewidth=2,label='Female Students')
plt.title('Student Marks wrt study hours',color='red',fontsize=20)
plt.xlabel('Student Marks',color='blue',fontsize=12)
plt.ylabel('Student Studied in Hours',color='blue',fontsize=12)
plt.xticks(color='green')
plt.yticks(color='green')
plt.grid(True)
plt.legend()
plt.show()
```

| Scatter Plot            | Line Plot            |
| ----------------------- | -------------------- |
| Shows relationship      | Shows trend          |
| Points only             | Points connected     |
| Correlation analysis    | Time-series analysis |
| Individual observations | Continuous data      |

| Scatter         | Bubble          |
| --------------- | --------------- |
| X and Y         | X,Y,Size        |
| Two variables   | Three variables |
| Same point size | Variable size   |
