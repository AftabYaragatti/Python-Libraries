### **Lineplot in Data Analytics**
#### **What is a Line Plot :**
##### A Line Plot is used to show the relationship between variables by connecting data points with lines.
##### **It is mainly used for:**
##### Trend Analysis
##### Time Series Analysis
##### Sales Analysis
##### Stock Market Analysis
##### Growth Analysis
##### **Example**
| Month | Sales |
| ----- | ----- |
| Jan   | 100   |
| Feb   | 120   |
| Mar   | 150   |
| Apr   | 180   |
| May   | 200   |
##### The points are connected using a line.
##### **Why Use Line Plot :**
##### Show trends over time
##### Time Series Analysis
##### Revenue Growth Analysis
##### Stock Market Analysis
##### Business Reporting
##### Forecasting
##### Machine Learning EDA
##### **Syntax :** sns.lineplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    style=None,
    size=None,
    palette=None,
    color=None,
    marker=None,
    markers=False,
    dashes=True,
    linewidth=2,
    alpha=None,
    estimator='mean',
    errorbar=('ci',95),
    legend='auto'
)
| X-axis      | Y-axis      | Valid?            |
| ----------- | ----------- | ----------------- |
| Numerical   | Numerical   | ✅ Yes             |
| Date/Time   | Numerical   | ✅ Most Common     |
| Categorical | Numerical   | ✅ Yes             |
| Numerical   | Categorical | ❌ No              |
| Categorical | Categorical | ❌ Not Recommended |

#### **dashes :**
##### Controls dashed lines.(True/False)
##### Seaborn uses different line styles:
##### Solid line (-)
##### Dashed line (--)
##### Dotted line (:)
##### Dash-dot line (-.)
##### dashes=True + linestyle=':'
##### These can conflict.
##### When you use: dashes=True
##### Seaborn automatically assigns different dash patterns.
##### Then: linestyle=':' tries to make all lines dotted.So choose one.

```python
sns.lineplot(data=df,x='Degree Yop',y='Degree Percentage',hue='Courses Taken',marker='D',dashes=True)
plt.title('Line Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.xlabel('Degree YOP',color='purple',fontsize=12)
plt.ylabel('Degree Percentage',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True)
plt.legend(loc='best',bbox_to_anchor=(1.05,1),)
plt.tight_layout()
plt.show()
```

#### **legend :**
##### Controls legend display.
##### **Values**
##### legend='auto'
##### legend='brief'
##### legend='full'
##### legend=False
##### **legend='auto' (Default)**
##### Seaborn automatically decides how much information to show.
##### For a small number of categories → shows all categories.
##### For many categories → may simplify the legend.
##### **legend='brief'**
##### Shows a summarized legend.
##### Mainly useful when the hue variable is numeric and contains many unique values.
##### **legend='full'**
##### Displays every unique value in the legend.
##### legend=False : Removes the legend completely.
| Value | Meaning                           | Best For           |
| ----- | --------------------------------- | ------------------ |
| auto  | Seaborn decides                   | General use        |
| brief | Shows fewer representative values | Large numeric data |
| full  | Shows all values                  | Detailed analysis  |
| False | No legend                         | Clean plots        |

```python
sns.lineplot(data=df,x='Degree Yop',y='Degree Percentage',hue='Courses Taken',marker='D',dashes=False,legend='full')
plt.title('Line Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.xlabel('Degree YOP',color='purple',fontsize=12)
plt.ylabel('Degree Percentage',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True)
plt.legend(loc='best',bbox_to_anchor=(1.05,1),)
plt.tight_layout()
plt.show()
```

#### **estimator :**
##### Aggregation function.When multiple observations exist for the same X-value.
##### Values : mean, median, sum, min & max

```python
import numpy as np
sns.lineplot(data=df,x='Degree Yop',y='Degree Percentage',hue='Courses Taken',marker='D',size='Admission Type',estimator=np.max)
plt.title('Line Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.xlabel('Degree YOP',color='purple',fontsize=12)
plt.ylabel('Degree Percentage',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True)
plt.legend(loc='best',bbox_to_anchor=(1.05,1),)
plt.tight_layout()
plt.show()
```

#### **errorbar :**
##### Displays uncertainty around estimates.
##### Values : errorbar=None, errorbar='sd', errorbar=('ci',95), errorbar=('ci',99)
##### **Example :** errorbar=('ci',95) -> Shows 95% Confidence Interval.

```python
import numpy as np
sns.lineplot(data=df,x='Degree Yop',y='Degree Percentage',hue='Courses Taken',marker='D',size='Admission Type',estimator=np.max,errorbar=('ci',95))
plt.title('Line Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.xlabel('Degree YOP',color='purple',fontsize=12)
plt.ylabel('Degree Percentage',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True)
plt.legend(loc='best',bbox_to_anchor=(1.05,1),)
plt.tight_layout()
plt.show()
```

#### **Example :**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df=pd.DataFrame({
    'Month':['Jan','Feb','Mar','Apr','May'],
    'Sales':[100,120,150,180,200],
    'Year':['2025','2025','2025','2025','2025'],
    'Profit':[10,15,20,25,30]
})

plt.figure(figsize=(10,6))

sns.lineplot(
    data=df,
    x='Month',
    y='Sales',
    hue='Year',
    marker='D',
    markers=True,
    linewidth=3,
    palette='Set2',
    alpha=0.8,
    legend='full'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)

plt.show()
```
