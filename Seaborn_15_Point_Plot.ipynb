# **Seaborn Library for Data Analytics**

#### **What is a Point Plot :**
##### A Point Plot is used to show the relationship between a categorical variable and a numerical variable using points connected by lines.
##### It displays the estimated value (mean by default) for each category.
##### **Definition :** A Point Plot visualizes aggregated statistics (usually mean) for categories using points and lines.
##### **Example Data :**
| Department | Salary |
| ---------- | ------ |
| IT         | 50000  |
| IT         | 55000  |
| HR         | 35000  |
| HR         | 37000  |
| Sales      | 45000  |
| Sales      | 48000  |
##### **Average Salary :**
##### IT = 52500
##### HR = 36000
##### Sales = 46500
##### **Why Use Point Plot :**
##### Compare category means
##### Compare trends across groups
##### Statistical analysis
##### Survey analysis
##### Business reporting
##### Machine Learning EDA
##### Better alternative to grouped bar plots
| Point Plot             | Bar Plot                |
| ---------------------- | ----------------------- |
| Uses points            | Uses bars               |
| Easier comparison      | Takes more space        |
| Better for trends      | Better for presentation |
| Highlights differences | Highlights magnitude    |
##### **Syntax :** sns.pointplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    order=None,
    hue_order=None,
    estimator='mean',
    errorbar=('ci',95),
    dodge=False,
    join=True,
    markers='o',
    linestyles='-',
    palette=None,
    color=None,
    scale=1,
    legend='auto'
)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
df=pd.read_csv('Employee_Dataset.csv')
df
```

##### **X :** Categorical Variable
##### **Y :** Numerical Variable

```python
sns.pointplot(data=df,x='Department',y='Experience_Years')
plt.title('Point Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Experience',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.5)
plt.tight_layout()
plt.show()
```

#### **hue :**
##### Creates separate colored points.

```python
sns.pointplot(data=df,x='Department',y='Experience_Years',hue='Gender')
plt.title('Point Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Experience',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.5)
plt.tight_layout()
plt.show()
```

#### **hue_order :**
##### It will change/rearrange the order of hue

```python
sns.pointplot(data=df,x='Department',y='Experience_Years',hue='Gender',hue_order=['Female','Male'])
plt.title('Point Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Experience',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.5)
plt.tight_layout()
plt.show()
```

#### **color :**

##### **Example :** color=green,red,blue,green,orange,black,purple,pink,gray,brown,cyan

```python
sns.pointplot(data=df,x='Department',y='Experience_Years',color='maroon')
plt.title('Point Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Experience',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.5)
plt.tight_layout()
plt.show()
```

#### **estimator :**
##### Function used to aggregate values.
##### **Values :** np.mean , np.median , np.sum , np.min , np.max

```python
import numpy as np
sns.pointplot(data=df,x='Department',y='Experience_Years',hue='Gender',estimator=np.sum)
plt.title('Point Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Experience',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.5)
plt.tight_layout()
plt.show()
```
