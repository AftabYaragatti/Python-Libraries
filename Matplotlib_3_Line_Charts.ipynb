# **Matplotlib Library for Data Analytics**

### **Line Charts in Matplotlib :** 
##### A Line Chart is used to show trends, patterns, growth, decrease, and changes over time or ordered data. OR A Line Chart connects data points using straight lines to represent continuous data.
##### **Syntax : plt.plot(x, y)**
##### **Different Types of Line Charts :**
| No | Type                  | Description                        |
| -- | --------------------- | ---------------------------------- |
| 1  | Single Line Plot      | One line                           |
| 2  | Multiple Line Plot    | Multiple lines                     |
| 3  | Line with Markers     | Data points visible                |
| 4  | Styled Line Plot      | Different line styles              |
| 5  | Colored Line Plot     | Different colors                   |
| 6  | Thick/Thin Line Plot  | Different line widths              |
| 7  | Dashed Line Plot      | Dashed lines                       |
| 8  | Dotted Line Plot      | Dotted lines                       |
| 9  | Step Line Plot        | Step-like graph                    |
| 10 | Error Line Plot       | Shows uncertainty/error            |
| 11 | Area Line Plot        | Filled line chart                  |
| 12 | Log Scale Line Plot   | Logarithmic axis                   |
| 13 | Time Series Line Plot | Dates on X-axis                    |
| 14 | Subplot Line Charts   | Multiple line charts in one figure |
| 15 | Animated Line Plot    | Dynamic line chart                 |

#### **1. Single Line Plot :**
##### A graph containing only one line.
##### **Syntax : plt.plot(x, y, color=None, linestyle='-', linewidth=1, marker=None, markersize=None, markerfacecolor=None, markeredgecolor=None, label=None, alpha=None)**
| Parameter       | Description            |
| --------------- | ---------------------- |
| x               | X-axis values          |
| y               | Y-axis values          |
| color           | Color of line          |
| linestyle       | Style of line          |
| linewidth       | Thickness of line      |
| marker          | Symbol at data points  |
| markersize      | Size of marker         |
| markerfacecolor | Fill color of marker   |
| markeredgecolor | Border color of marker |
| label           | Name used in legend    |
| alpha           | Transparency (0 to 1)  |
##### **Parameter Details with Default Value :**
| Parameter         | Default Value              | Description                           |
| ----------------- | -------------------------- | ------------------------------------- |
| `x`               | Required                   | X-axis data                           |
| `y`               | Required                   | Y-axis data                           |
| `color`           | `None`                     | Uses Matplotlib's default color cycle |
| `linestyle`       | `'-'`                      | Solid line                            |
| `linewidth`       | `1.5`                      | Line thickness                        |
| `marker`          | `None`                     | No marker displayed                   |
| `markersize`      | `6`                        | Marker size                           |
| `markerfacecolor` | `None`                     | Same as line color                    |
| `markeredgecolor` | `None`                     | Same as line color                    |
| `label`           | `None`                     | No legend label                       |
| `alpha`           | `None` (equivalent to 1.0) | Fully opaque                          |

```python
import matplotlib.pyplot as plt
years = [2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024,2025]
growth = [15,19,22,26,28,29,32,35,38,34,21,22,28,34,38,67]
plt.plot(years,growth,color='green',linestyle='--',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='Sales Report')
plt.title('Sales : Growth VS Years',color='red',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=20)
plt.xlabel('Years',color='blue',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=15)
plt.ylabel('Growth',color='blue',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=15)
plt.grid(True)
plt.legend()
plt.show()
```

#### **2. Multiple Line Plot :**
##### More than one line in the same graph for comparison.
##### **Syntax 1: plt.plot(x1, y1, color=None, linestyle='-', linewidth=1, marker=None, markersize=None, markerfacecolor=None, markeredgecolor=None, label=None, alpha=None)**

```python
import matplotlib.pyplot as plt
years_tcs = [2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024,2025]
growth_tcs = [15,19,22,26,28,29,32,35,38,34,21,22,28,34,38,67]
years_ibm = [2004,2005,2006,2007,2008,2009,2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024,2025]
growth_ibm = [17,23,25,26,28,31,35,39,42,46,48,49,52,55,58,54,31,26,28,34,47,86]
plt.plot(years_tcs,growth_tcs,color='green',linestyle='--',linewidth=2,marker='o',markeredgecolor='green',markerfacecolor='red',label='TCS Sales Report')
plt.plot(years_ibm,growth_ibm,color='blue',linestyle='--',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='IBM Sales Report')
plt.title('Sales IBM & TCS : Growth VS Years',color='red',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=20)
plt.xlabel('Years',color='blue',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=15)
plt.ylabel('Growth',color='blue',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=15)
plt.grid(True)
plt.legend()
plt.show()
```

#### **10. Error Line Plot :**
##### plt.errorbar() is used to plot data points along with error bars that represent uncertainty, variability, standard deviation, confidence intervals, measurement errors, etc.
##### **Syntax :** plt.errorbar(
    x,                      # X-axis values
    y,                      # Y-axis values
    yerr=None,              # Error in Y direction
    xerr=None,              # Error in X direction
    fmt='',                 # Format string
    ecolor=None,            # Error bar color
    elinewidth=None,        # Error bar thickness
    capsize=0,              # Cap length
    capthick=None,          # Cap thickness
    barsabove=False,        # Draw bars above markers
    lolims=False,           # Lower limit arrows
    uplims=False,           # Upper limit arrows
    xlolims=False,          # X lower limit arrows
    xuplims=False,          # X upper limit arrows
    errorevery=1,           # Show error bars frequency
    alpha=None,             # Transparency
    label=None,             # Legend label
    color=None,             # Marker/line color
    linestyle='None',       # Line style
    linewidth=None,         # Line thickness
    marker=None,            # Marker style
    markersize=None         # Marker size )

```python
import matplotlib.pyplot as plt
years = [2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024,2025]
growth = [15,19,22,26,28,29,32,35,38,34,21,22,28,34,38,67]
plt.plot(years,growth,color='green',linestyle='--',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='Sales Report')
plt.title('Sales : Growth VS Years',color='red',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=20)
plt.xlabel('Years',color='blue',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=15)
plt.ylabel('Growth',color='blue',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=15)
plt.grid(True)
plt.legend()
plt.errorbar(years, growth)
plt.show()
```

| `ncols`   | `1`     | Number of columns |
| `figsize` | None    | Figure size       |
| `sharex`  | False   | Share X-axis      |
| `sharey`  | False   | Share Y-axis      |
| `dpi`     | 100     | Figure resolution |
| `squeeze` | True    | Return axes shape |


##### **Example 1: Two Vertical Charts**

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(2,1)
ax[0].plot([1,2,3],[10,20,30])
ax[1].plot([1,2,3],[30,20,10])
plt.show()
```

##### **Example 2: Two Horizontal Charts**

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1,2)
ax[0].plot([1,2,3],[10,20,30])
ax[1].plot([1,2,3],[30,20,10])
plt.show()
```

##### **Example 3: 2×2 Subplots**

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(2,2)
ax[0,0].plot([1,2,3],[10,20,30])
ax[0,1].plot([1,2,3],[30,20,10])
ax[1,0].plot([1,2,3],[5,15,25])
ax[1,1].plot([1,2,3],[25,15,5])
plt.show()
```

##### **fig, ax = plt.subplots(2,1)**
##### ax[0].plot()
##### ax[1].plot()

```python
import matplotlib.pyplot as plt
years = [2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024,2025]
growth = [15,19,22,26,28,29,32,35,38,34,21,22,28,34,38,67]
fig, ax = plt.subplots(2,1)
ax[0].plot(years,growth,color='green',linestyle='--',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='Sales Report')
ax[1].plot(years,growth,color='green',linestyle='--',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='Sales Report')
plt.title('Sales : Growth VS Years',color='red',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=20)
plt.xlabel('Years',color='blue',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=15)
plt.ylabel('Growth',color='blue',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=15)
plt.grid(True)
plt.legend()
plt.show()
```

#### **figsize :**
##### Controls figure width and height.

```python
import matplotlib.pyplot as plt
years = [2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024,2025]
growth = [15,19,22,26,28,29,32,35,38,34,21,22,28,34,38,67]
fig, ax = plt.subplots(2,2,figsize=(14,6))
ax[0,0].plot(years,growth,color='green',linestyle='--',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='Sales Report')
ax[0,1].plot(years,growth,color='green',linestyle=':',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='Sales Report')
ax[1,0].plot(years,growth,color='green',linestyle='-.',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='Sales Report')
ax[1,1].plot(years,growth,color='green',linestyle='-',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='Sales Report')
plt.show()
```

#### **sharex :**
##### All subplots share the same X-axis.

```python
import matplotlib.pyplot as plt
years = [2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024,2025]
growth = [15,19,22,26,28,29,32,35,38,34,21,22,28,34,38,67]
fig, ax = plt.subplots(2,2,figsize=(14,6),sharex=True)
ax[0,0].plot(years,growth,color='green',linestyle='--',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='Sales Report')
ax[0,1].plot(years,growth,color='green',linestyle=':',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='Sales Report')
ax[1,0].plot(years,growth,color='green',linestyle='-.',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='Sales Report')
ax[1,1].plot(years,growth,color='green',linestyle='-',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='Sales Report')
plt.show()
```
