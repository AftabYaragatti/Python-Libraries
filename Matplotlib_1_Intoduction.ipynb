# **Matplotlib Library for Data Analytics**

### **Intorduction**

##### **1. What is Matplotlib?**
##### Matplotlib is a powerful Python library used for data visualization. 
##### It helps us represent data graphically using charts and plots, making it easier to understand patterns, trends, and relationships in data.
##### It was developed by John D. Hunter in 2003.
##### Matplotlib is a Python library used to create graphs, charts, and visual representations of data.

##### **Examples of Graphs Created Using Matplotlib**
##### 1.Line Chart
##### 2.Bar Chart
##### 3.Pie Chart
##### 4.Histogram
##### 5.Scatter Plot
##### 6.Area Chart
##### 7.Box Plot

##### **Why Visualization?** Consider the following data:
| Month | Sales |
| ----- | ----- |
| Jan   | 100   |
| Feb   | 150   |
| Mar   | 200   |

##### By looking at numbers alone, understanding growth is difficult. A graph instantly shows the increasing trend.

##### **2. Why Use Matplotlib?**
##### Matplotlib is one of the most widely used visualization libraries in Data Analytics, Data Science, and Machine Learning.
##### **Advantages**
##### **1. Easy to Learn :** Matplotlib provides simple and beginner-friendly functions for creating visualizations.
##### **2. Creates Professional Charts :** It helps generate high-quality and publication-ready graphs and charts.
##### **3. Highly Customizable :** Almost every part of a graph can be customized, including colors, labels, fonts, and styles.
##### **4. Supports Multiple Chart Types :** It can create various charts such as line, bar, pie, scatter, histogram, and more.
##### **5. Works with Other Libraries :** Matplotlib integrates seamlessly with NumPy, Pandas, and Seaborn for efficient data analysis and visualization.

##### **4. Importing Matplotlib**
##### Matplotlib contains many modules. The most commonly used module is:
##### matplotlib.pyplot
##### **Standard Import**
##### **import matplotlib.pyplot as plt**
##### where matplotlib is Main library & pyplot is Plotting module & plt is Alias (shortcut name)
##### **5. Understanding Pyplot**
##### **What is Pyplot?** 
##### Pyplot is a module inside Matplotlib that provides functions similar to MATLAB. It helps create, customize, and display plots easily.
##### Think of pyplot as:
##### A drawing tool that helps create graphs with simple commands.
##### **Common Pyplot Functions**
| Function      | Purpose       |
| ------------- | ------------- |
| plt.plot()    | Line graph    |
| plt.bar()     | Bar chart     |
| plt.scatter() | Scatter plot  |
| plt.hist()    | Histogram     |
| plt.pie()     | Pie chart     |
| plt.title()   | Add title     |
| plt.xlabel()  | X-axis label  |
| plt.ylabel()  | Y-axis label  |
| plt.legend()  | Add legend    |
| plt.grid()    | Add grid      |
| plt.show()    | Display graph |

```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.plot(x,y)
plt.show()
```

| Component            | Purpose                            |
| -------------------- | ---------------------------------- |
| Figure               | Complete window containing plots   |
| Axes                 | Plotting area where graph is drawn |
| Axis                 | X and Y scales of graph            |
| Plot Line            | Draws data on graph                |
| Labels               | Names of X and Y axes              |
| Title                | Heading of graph                   |
| Legend               | Identifies plotted data            |
| Grid                 | Reference lines for reading values |
| Ticks                | Axis markings and values           |
| Spines               | Borders around plot                |
| Text                 | Custom text on graph               |
| Annotation           | Text with arrow pointing to data   |
| Plot Area Background | Background color inside plot       |
| Figure Background    | Background color of entire figure  |

#### **Plot Components :**
| No.    | Component                | Theory / Notes                                                                                 | Full Syntax with Parameters                                                                                               | Important Parameters                                                                                                                                                                    |
| ------ | ------------------------ | ---------------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------------------------------------- | --------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **1**  | **Figure**               | The complete canvas/window that holds one or more plots (Axes). It is the top-level container. | `fig = plt.figure(figsize=(8,5), dpi=100, facecolor='white', edgecolor='black')`                                          | `figsize=(w,h)` → size in inches<br>`dpi` → resolution<br>`facecolor` → background color<br>`edgecolor` → border color                                                                  |
| **2**  | **Axes**                 | The actual plotting area where graphs are drawn. Most plotting functions belong to Axes.       | `ax = fig.add_subplot(111)`<br>`fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(8,5))`                                  | `nrows` → rows<br>`ncols` → columns<br>`sharex` → share x-axis<br>`sharey` → share y-axis                                                                                               |
| **3**  | **Axis**                 | Represents X-axis and Y-axis scales, limits, and tick positions.                               | `ax.set_xlim(left=0, right=10)`<br>`ax.set_ylim(bottom=0, top=100)`<br>`ax.set_xscale('linear')`                          | `left,right` → x-range<br>`bottom,top` → y-range<br>`linear,log,symlog,logit`                                                                                                           |
| **4**  | **Plot Line**            | Draws line charts connecting data points. Most commonly used plotting function.                | `ax.plot(x, y, color='blue', linestyle='-', linewidth=2, marker='o', markersize=8, label='Sales', alpha=1.0)`             | `color` → line color<br>`linestyle` → line style<br>`linewidth` → thickness<br>`marker` → point symbol<br>`markersize` → marker size<br>`label` → legend text<br>`alpha` → transparency |
| **5**  | **X Label**              | Displays name/description of X-axis.                                                           | `ax.set_xlabel(xlabel='Months', fontsize=12, color='black', labelpad=10, loc='center')`                                   | `xlabel` → text<br>`fontsize` → size<br>`color` → color<br>`labelpad` → spacing<br>`loc` → position                                                                                     |
| **6**  | **Y Label**              | Displays name/description of Y-axis.                                                           | `ax.set_ylabel(ylabel='Sales', fontsize=12, color='black', labelpad=10, loc='center')`                                    | `ylabel` → text<br>`fontsize` → size<br>`color` → color<br>`labelpad` → spacing                                                                                                         |
| **7**  | **Title**                | Main heading of graph. Helps identify chart purpose.                                           | `ax.set_title(label='Monthly Sales', fontsize=16, color='red', pad=20, loc='center')`                                     | `label` → title text<br>`fontsize` → size<br>`color` → color<br>`pad` → distance<br>`loc` → left/center/right                                                                           |
| **8**  | **Legend**               | Explains plotted lines, bars, or datasets.                                                     | `ax.legend(loc='best', fontsize=10, title='Data', frameon=True, shadow=False)`                                            | `loc` → position<br>`fontsize` → text size<br>`title` → legend title<br>`frameon` → border<br>`shadow` → shadow effect                                                                  |
| **9**  | **Grid**                 | Displays horizontal and vertical reference lines for easier reading.                           | `ax.grid(visible=True, color='gray', linestyle='--', linewidth=0.5, alpha=0.7)`                                           | `visible` → show/hide<br>`color` → grid color<br>`linestyle` → style<br>`linewidth` → thickness<br>`alpha` → transparency                                                               |
| **10** | **Ticks**                | Marks and labels on X-axis and Y-axis.                                                         | `ax.tick_params(axis='both', which='major', labelsize=10, labelcolor='blue', rotation=45, direction='in')`                | `axis` → x/y/both<br>`which` → major/minor<br>`labelsize` → text size<br>`rotation` → rotate labels<br>`direction` → in/out                                                             |
| **11** | **Spines**               | Border lines around plotting area.                                                             | `ax.spines['top'].set_visible(False)`<br>`ax.spines['bottom'].set_color('red')`<br>`ax.spines['bottom'].set_linewidth(2)` | `top`<br>`bottom`<br>`left`<br>`right`<br>`set_visible()`<br>`set_color()`<br>`set_linewidth()`                                                                                         |
| **12** | **Text**                 | Places custom text anywhere on graph.                                                          | `ax.text(x=3, y=20, s='Sales Peak', fontsize=12, color='red', ha='center', va='center')`                                  | `x,y` → position<br>`s` → text<br>`fontsize` → size<br>`color` → color<br>`ha` → horizontal align<br>`va` → vertical align                                                              |
| **13** | **Annotation**           | Text with arrow pointing to a specific data point.                                             | `ax.annotate(text='Highest', xy=(5,30), xytext=(4,40), arrowprops=dict(facecolor='black', shrink=0.05), fontsize=12)`     | `text` → annotation text<br>`xy` → target point<br>`xytext` → text location<br>`arrowprops` → arrow settings                                                                            |
| **14** | **Plot Area Background** | Changes background color inside graph area (Axes).                                             | `ax.set_facecolor(color='lightyellow')`                                                                                   | `color` → background color                                                                                                                                                              |
| **15** | **Figure Background**    | Changes background color of entire figure/canvas.                                              | `fig.set_facecolor(color='lightblue')`                                                                                    | `color` → background color                                                                                                                                                              |
#### **Important Values :**
| Feature              | Common Values                                                          |
| -------------------- | ---------------------------------------------------------------------- |
| `linestyle`          | `'-'`, `'--'`, `':'`, `'-.'`                                           |
| `marker`             | `'o'`, `'s'`, `'^'`, `'*'`, `'D'`, `'+'`, `'x'`                        |
| `loc` (Legend/Title) | `'best'`, `'upper right'`, `'upper left'`, `'lower right'`, `'center'` |
| `axis`               | `'x'`, `'y'`, `'both'`                                                 |
| `which`              | `'major'`, `'minor'`, `'both'`                                         |
| `direction`          | `'in'`, `'out'`, `'inout'`                                             |
| `ha`                 | `'left'`, `'center'`, `'right'`                                        |
| `va`                 | `'top'`, `'center'`, `'bottom'`                                        |
| `scale`              | `'linear'`, `'log'`, `'symlog'`, `'logit'`                             |


```python
import matplotlib.pyplot as plt
years = [2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024,2025]
growth = [15,19,22,26,28,29,32,35,38,34,21,22,28,34,38,67]
plt.plot(years,growth,color='green',linestyle='--',linewidth=2,marker='o',markeredgecolor='blue',markerfacecolor='red',label='Sales Report')
plt.legend()
plt.title('Sales : Growth VS Years',color='red',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=20)
plt.xlabel('Years',color='blue',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=15)
plt.ylabel('Growth',color='blue',fontweight='bold',fontstyle='italic',fontfamily='Times New Roman',fontsize=15)
plt.grid(True)
plt.show()
```
