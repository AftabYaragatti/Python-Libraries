# **Seaborn Library for Data Analytics**

#### **What is KDE Plot :**
##### KDE stands for Kernel Density Estimation.
##### A KDE Plot is used to visualize the probability density distribution of numerical data using a smooth curve.
##### **Definition :** A KDE Plot is a smooth version of a Histogram that shows how data is distributed.

| Histogram            | KDE Plot                         |
| -------------------- | -------------------------------- |
| Uses bars            | Uses smooth curve                |
| Shows frequency      | Shows density                    |
| Discrete appearance  | Continuous appearance            |
| Easier to understand | Better statistical visualization |

#### **Why Use KDE Plot :**
##### Understand Data Distribution
##### Detect Skewness
##### Compare Multiple Groups
##### Identify Peaks
##### Machine Learning EDA
##### Statistical Analysis
##### Alternative to Histogram

##### **Syntax :** sns.kdeplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    fill=False,
    multiple='layer',
    common_norm=True,
    bw_adjust=1,
    cumulative=False,
    palette=None,
    color=None,
    alpha=None,
    linewidth=2,
    linestyle='-',
    legend=True
)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('Employee_Cleaned_Dataset.csv')
df
```

##### **x :** Numerical variable displayed on X-axis.
##### **y :** Creates a 2D KDE Plot.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10,4))
sns.kdeplot(data=df,x='Age')
plt.title('KDE Plot',fontsize=20,color='red',fontstyle='italic',fontfamily='Times New Roman',fontweight='bold')
plt.xlabel('Age',fontsize=15,color='purple',fontfamily='Times New Roman')
plt.ylabel('Density',fontsize=15,color='purple',fontfamily='Times New Roman')
plt.xticks(color='blue',fontsize=14,rotation=0)
plt.yticks(color='blue',fontsize=14,rotation=0)
plt.grid(True,axis='y',linestyle='--',alpha=0.7)
plt.tight_layout()
plt.show()
```
