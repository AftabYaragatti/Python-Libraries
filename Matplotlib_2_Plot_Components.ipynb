# **Matplotlib Library for Data Analytics**

#### **plt.plot() in Matplotlib :**
##### **Definition :** plt.plot() is used to create a line graph by plotting points on X-axis and Y-axis and connecting them with lines.It is the most commonly used function in Matplotlib.
##### **Syntax : plt.plot(x, y, color=None, linestyle='-', linewidth=1, marker=None, markersize=None, markerfacecolor=None, markeredgecolor=None, label=None, alpha=None)**
| Parameter       | Description            |
| --------------- | ---------------------- |
| x               | X-axis values          |
| y               | Y-axis values          |
| color           | Color of line          |
| linestyle       | Style of line          |
| linewidth       | Thickness of line      |
| marker          | Symbol at data points  |
| markersize      | Size of marker         |
| markerfacecolor | Fill color of marker   |
| markeredgecolor | Border color of marker |
| label           | Name used in legend    |
| alpha           | Transparency (0 to 1)  |
##### **Parameter Details with Default Value :**
| Parameter         | Default Value              | Description                           |
| ----------------- | -------------------------- | ------------------------------------- |
| `x`               | Required                   | X-axis data                           |
| `y`               | Required                   | Y-axis data                           |
| `color`           | `None`                     | Uses Matplotlib's default color cycle |
| `linestyle`       | `'-'`                      | Solid line                            |
| `linewidth`       | `1.5`                      | Line thickness                        |
| `marker`          | `None`                     | No marker displayed                   |
| `markersize`      | `6`                        | Marker size                           |
| `markerfacecolor` | `None`                     | Same as line color                    |
| `markeredgecolor` | `None`                     | Same as line color                    |
| `label`           | `None`                     | No legend label                       |
| `alpha`           | `None` (equivalent to 1.0) | Fully opaque                          |
##### **Marker style :**
| Marker | Shape          |
| ------ | -------------- |
| 'o'    | Circle         |
| '*'    | Star           |
| '.'    | Point          |
| '+'    | Plus           |
| 'x'    | Cross          |
| 's'    | Square         |
| 'D'    | Diamond        |
| '^'    | Triangle Up    |
| 'v'    | Triangle Down  |
| '<'    | Left Triangle  |
| '>'    | Right Triangle |
| 'p'    | Pentagon       |
| 'h'    | Hexagon        |
##### **Example :**

```python
import matplotlib.pyplot as plt
x=[2,4,5,7,9,10]
y=[1,3,6,9,12,13]
print(x,y)
plt.plot(x,y)
```

#### **1. Figure :**
##### A Figure is the top-level container or canvas in Matplotlib that holds all plot elements such as Axes, Titles, Labels, Legends, and Grids. In simple words, Figure is the complete window on which graphs are drawn.
##### **Syntax : plt.figure(num=None, figsize=None, dpi=None, facecolor=None, edgecolor=None, frameon=True, clear=False)**
##### **Parameters :**
| Parameter   | Description                          |
| ----------- | ------------------------------------ |
| `num`       | Figure number or name                |
| `figsize`   | Width and height of figure in inches |
| `dpi`       | Resolution (Dots Per Inch)           |
| `facecolor` | Background color of figure           |
| `edgecolor` | Border color of figure               |
| `frameon`   | Show or hide figure frame            |
| `clear`     | Clears existing figure if True       |
##### **Parameters Details with Defualt Value :**
| Parameter   | Default Value                               | Description                    |
| ----------- | ------------------------------------------- | ------------------------------ |
| `num`       | `None`                                      | Figure number or name          |
| `figsize`   | `rcParams["figure.figsize"]` → `(6.4, 4.8)` | Width and height in inches     |
| `dpi`       | `rcParams["figure.dpi"]` → `100.0`          | Resolution (Dots Per Inch)     |
| `facecolor` | `rcParams["figure.facecolor"]` → `'white'`  | Background color               |
| `edgecolor` | `rcParams["figure.edgecolor"]` → `'white'`  | Border color                   |
| `frameon`   | `True`                                      | Displays figure frame          |
| `clear`     | `False`                                     | Does not clear existing figure |

##### **NOTE :** If we are using plt.plot() and plt.figure() make sure that we have to use first plt.figure() then followed by plt.plot() in order to get the visuals in figure component also

##### **num :**
##### **Syntax : plt.figure(num=None)**
##### num = Figure number or name ( 1 or 'Sales')

#### **2. Axes :**
##### **Definition :** Axes is the area inside a Figure where the actual graph is drawn.
##### **It contains:** X-Axis, Y-Axis, Plot Data, Labels, Title, Legend, Grid
##### **or** Axes is the rectangular plotting area where data is displayed.
##### **Real-Life Analogy :**
##### 📝 Figure = Notebook Page
##### 📦 Axes = Box drawn on the page
##### 📈 Plot = Graph inside the box
##### **Syntax :**
##### **Method 1 : ax = plt.axes()**
##### **Method 2 (Most Common) : fig, ax = plt.subplots()**
##### **Default Syntax : ax = plt.axes(arg=None, facecolor=None, frameon=True, sharex=None, sharey=None, label='')**
##### **Important Parameters :**
| Parameter   | Default Value  | Description              |
| ----------- | -------------- | ------------------------ |
| `facecolor` | `None` (white) | Background color of axes |
| `frameon`   | `True`         | Show axes border/frame   |
##### Axes is the rectangular area where plt.plot() draws the graph. When you create a figure using plt.figure(), Matplotlib automatically creates an Axes object when you call plt.plot().

##### **facecolor :**
##### **Syntax : plt.axes(facecolor=None)**
##### facecolor = Background color of axes

```python
import matplotlib.pyplot as plt
years=[2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020]
growth=[45,52,59,63,68,75,80,85,89,90,106]
plt.figure(num='Sales Report',figsize=(10,8),dpi=70,facecolor='cyan',edgecolor='purple',frameon=True,clear=False)
plt.axes(facecolor='lightyellow')
plt.plot(years,growth,color='c',linestyle='-',linewidth=3,marker='o',markersize=8,markerfacecolor='r',markeredgecolor='b',label='Sales')
```

| --------- | ------- | --------------------------------- |
| `left`    | `None`  | Left limit of X-axis              |

#### **13. Plot Area Background :**
##### **13.1 ax.set_facecolor()** is used to set the background color of the Axes (plot area). It changes only the area where data is plotted.
##### For Changing Figure Background we use : **fig.set_facecolor('cyan')** It Changes outer area.
##### **Plot Area Background**
##### **ax.set_facecolor('lightyellow') :** Changes inner plotting area.
##### **Named Colors :** yellow, cyan, lightblue, lightgreen, pink
##### **Short Colors :** r=red, g=green, b=blue, c=cyan, m=magenta, y=yellow, k=black, w=white
##### **Hex Colors :** #FFFF00, #ADD8E6, #90EE90
##### **RGB Tuple :** Values range from 0 to 1.(1,0,0)=Red, (0,1,0)=Green, (0,0,1)=Blue
##### **RGBA Tuple :** (1,1,0,0.5) Semi-transparent yellow
##### * R = Red
##### * G = Green
##### * B = Blue
##### * A = Alpha (Transparency)
| Method                | Purpose                     |
| --------------------- | --------------------------- |
| `ax.set_facecolor()`  | Set plot area background    |
| `ax.get_facecolor()`  | Get plot area background    |

```python
import matplotlib.pyplot as plt
years=[2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020]
growth=[45,52,59,63,68,75,80,85,89,90,106]

fig, ax = plt.subplots(figsize=(10,8), dpi=70)
ax.set_facecolor('purple')
print(ax.get_facecolor())

plt.plot(years,growth,color='purple',linestyle='--',linewidth=1.5,marker='o',markersize=8,markerfacecolor='purple',markeredgecolor='black',markeredgewidth=5,label='NEW SALES',alpha=0.6)
plt.xlabel('Years',color='purple',fontsize=15)
plt.ylabel('Growth',color='purple',fontsize=15)
plt.title('Sales',fontfamily='Times New Roman',fontweight='bold',color='blue',fontsize=30,rotation=0)
plt.grid(True)
plt.show()
```

#### **14. Figure Background :**
##### **14.1 fig.set_facecolor() :**
##### fig.set_facecolor()` is used to set the **background color of the entire Figure (canvas)**.
##### The Figure is the outer container that contains all Axes (plots), titles, legends, and other elements.
##### **Syntax : fig.set_facecolor(color)**
| Parameter | Default | Description                    |
| --------- | ------- | ------------------------------ |
| `color`   | None    | Background color of the figure |
##### Get Current Figure Background : print(fig.get_facecolor()) It returns the output in RGBA format
| Method                | Purpose                     |
| --------------------- | --------------------------- |
| `fig.set_facecolor()` | Set figure background color |
| `fig.get_facecolor()` | Get figure background color |
##### **Difference Between Figure and Axes Background :**
| Figure Background     | Axes Background       |
| --------------------- | --------------------- |
| Entire canvas         | Plot area only        |
| `fig.set_facecolor()` | `ax.set_facecolor()`  |
| Outer region          | Inner plotting region |

```python
import matplotlib.pyplot as plt
years=[2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020]
growth=[45,52,59,63,68,75,80,85,89,90,106]

fig, ax = plt.subplots(figsize=(10,8), dpi=70)
ax.set_facecolor('purple')
print(ax.get_facecolor())
fig.set_facecolor('cyan')
print(fig.get_facecolor())

plt.plot(years,growth,color='purple',linestyle='--',linewidth=1.5,marker='o',markersize=8,markerfacecolor='purple',markeredgecolor='black',markeredgewidth=5,label='NEW SALES',alpha=0.6)
plt.xlabel('Years',color='purple',fontsize=15)
plt.ylabel('Growth',color='purple',fontsize=15)
plt.title('Sales',fontfamily='Times New Roman',fontweight='bold',color='blue',fontsize=30,rotation=0)
plt.grid(True)
plt.show()
```
