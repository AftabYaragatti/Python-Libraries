# **Matplotlib Library for Data Analytics**

### **Imshow(Heatmap) in Matplotlib**

#### **What is imshow() :** imshow() stands for: Image Show
##### It is used to display:
##### Images
##### 2D Arrays
##### Matrices
##### Heatmaps
##### Pixel Data
##### In Data Analytics, it is commonly used to create Heatmaps.
##### **What is a Heatmap :** A Heatmap uses colors to represent values.
##### **Example Matrix:**
##### data = [
    [10,20,30],
    [40,50,60],
    [70,80,90]
]
##### **Heatmap Concept:**
##### 10  → Light Color
##### 50  → Medium Color
##### 90  → Dark Color
##### Instead of numbers, colors show intensity.
##### **Real-Life Examples :**
##### Student Marks
##### Math      90
##### Science   75
##### English   40
##### **Heatmap:**
##### 90 → Dark
##### 75 → Medium
##### 40 → Light
##### **Temperature Map**
##### 20°C → Blue
##### 40°C → Yellow
##### 60°C → Red
##### **Correlation Matrix**
##### Age vs Salary
##### Salary vs Experience
##### Very common in Data Analytics and Machine Learning.
##### **Syntax :** plt.imshow(
    X,
    cmap=None,
    aspect=None,
    interpolation=None,
    alpha=None,
    vmin=None,
    vmax=None,
    origin=None,
    extent=None
)

```python
import matplotlib.pyplot as plt
data = [
    [10,20,30],
    [40,50,60],
    [70,80,90]
]
plt.imshow(data)
plt.show()
```

```python
import matplotlib.pyplot as plt
marks = [
    [85, 90, 78],   # Student 1
    [70, 88, 92],   # Student 2
    [95, 80, 85]    # Student 3
]
plt.imshow(marks)
plt.title("Student Marks Heatmap")
plt.xlabel("Subjects")
plt.ylabel("Students")
plt.colorbar(label="Marks")
plt.show()
```

#### **cmap :**
##### Color map. Controls colors.
| Value   | Theme             |
| ------- | ----------------- |
| viridis | Blue-Green-Yellow |
| plasma  | Purple-Yellow     |
| inferno | Dark-Yellow       |
| magma   | Dark-Pink         |
| cool    | Cyan-Magenta      |
| hot     | Black-Red-Yellow  |
| gray    | Grayscale         |
| jet     | Rainbow           |

```python
import matplotlib.pyplot as plt
marks = [
    [85, 90, 78],   # Student 1
    [70, 88, 92],   # Student 2
    [95, 80, 85]    # Student 3
]
plt.imshow(marks,cmap='jet')
plt.title("Student Marks Heatmap")
plt.xlabel("Subjects")
plt.ylabel("Students")
plt.colorbar(label="Marks")
plt.show()
```

#### **extent :**
##### Controls axis ranges.
##### extent=[xmin,xmax,ymin,ymax]

```python
import matplotlib.pyplot as plt
marks = [
    [85, 90, 78],   # Student 1
    [70, 88, 92],   # Student 2
    [95, 80, 85]    # Student 3
]
plt.imshow(marks,cmap='jet',extent=[0,15,0,15])
plt.title("Student Marks Heatmap")
plt.xlabel("Subjects")
plt.ylabel("Students")
plt.colorbar(label="Marks")
plt.show()
```

#### **Colorbar :**
##### Almost always used with heatmaps.
##### plt.colorbar()
##### Colorbar => Almost always used with heatmaps : plt.colorbar()

```python
import matplotlib.pyplot as plt
sales = [
    [84,45,55,51],
    [94,51,56,59],
    [84,34,55,37]
]
plt.imshow(
    sales,
    cmap='viridis'
)
plt.colorbar(label='Sales')
plt.title('Sales Heatmap')
plt.show()
```

| imshow()             | contourf()             |
| -------------------- | ---------------------- |
| Grid-based           | Contour-based          |
| Cell colors          | Filled contour regions |
| Matrix visualization | Continuous surface     |

| imshow()     | pcolormesh()   |
| ------------ | -------------- |
| Regular grid | Irregular grid |
| Faster       | More flexible  |
