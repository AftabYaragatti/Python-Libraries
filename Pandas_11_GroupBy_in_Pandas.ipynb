# **Pandas Library**
### **Group By in Pandas**

##### **Group By :**
##### It is a DataFrame method which is used to group the records based on given columns. Group By Internally it performs boolean indexing & fetchs values based on unique values from a given columns.
##### After performing group by internally it creates dictionary where all the unique values of grouped columns act as keys and values are indexs of records.
##### **General Syntax : df.groupby('col_name')**

```python
import pandas as pd
import numpy as np
df=pd.read_csv("Employee_Cleaned_Dataset.csv")
df.drop(columns='Unnamed: 0', inplace=True)
df
```

```python
df.groupby('Department')
```

##### Multiple Column Grouping :
##### **Syntax : df.groupby(['col_name','col_name'])**

```python
df.groupby(['Department','City'])
```

##### Group by with indexing syntax :
##### **Syntax : df.groupby('col_name')['col_name']**

```python
df.groupby('Department')['Salary']
```

##### Group by with aggregation function :
##### **Syntax : df.groupby('col_name')['col_name'].fun()**
##### Where fun() includes sum(),count(),min(),max()

##### **Ex1. Group By Department and Calculate Total Salary**

```python
df.groupby('Department')['Salary'].sum()
```

##### **Ex2. Group By Gender and Calculate Total Salary**

```python
df.groupby('Gender')['Salary'].sum()
```

##### **Ex3. Group By Department and Calculate Average Salary**

```python
df.groupby('Department')['Salary'].mean()
```

##### **Ex4. Count Employees in Each Department**

```python
df.groupby('Department')['Employee Name'].count()
```

##### **Ex5. Find Maximum Salary in Each Department**

```python
df=df.groupby('Department')['Salary'].max()
s = df.to_frame()
s
```

##### **Ex6. Find Minimum Salary in Each Department**

```python
df.groupby('Department')['Salary'].min()
```

##### Apply multiple aggregation function :
##### **Syntax : df.groupby('col_name')['col_name'].agg(['fname','fname',...])**

```python
df.groupby('Department')['Salary'].agg(['sum', 'mean', 'max', 'min'])
```

```python
df.groupby('Gender')['Salary'].agg(['sum', 'mean', 'max', 'min'])
```

##### **GroupBy multiple Departments**

```python
df.groupby(['Department', 'Gender'])['Salary'].mean()
```

```python
df.groupby(['Department', 'Age'])['Salary'].mean()
```

```python
df.groupby(['Gender','Department'])['Salary'].mean()
```

```python
df.groupby(['Gender','Department','City'])['Salary'].mean()
```

```python
df.groupby(['Gender','Department'])['Employee Name'].count()
```

```python
df.groupby(['Gender','City'])['Employee Name'].count()
```

```python
df.groupby(['Gender','Department','City'])['Employee Name'].count()
```

##### **Fetching all the groups :**
##### To fetch all the groups we use group by attribute.
##### **Syntax : variable = df.groupby('col')**
##### **variable.groups**
##### Here group returns dictionary where keys are unique values given column & values are row index. 

```python
det=df.groupby('Department')
det.groups
```

```python
det=df.groupby('Gender')
det.groups
```

```python
det=df.groupby('City')
det.groups
```

##### To fetch single group separately we use group by internal methods.
##### **Syntax : variable = df.groupby('col')**
##### **variable.get_group('key')**
##### **key:** The actual value present in the grouped column

```python
det=df.groupby('City')
det.get_group('Bangalore')
```

```python
det=df.groupby('City')
det.get_group('Belagavi')
```

```python
det=df.groupby('Gender')
det.get_group('Male')
```

```python
det=df.groupby('Department')
det.get_group('IT')
```

##### **Convert Grouped Result to DataFrame**

```python
df.groupby('Department')['Salary'].mean().reset_index()
```
