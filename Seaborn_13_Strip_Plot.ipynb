# **Seaborn Library for Data Analytics**

#### **What is a Strip Plot :**
##### A Strip Plot is used to visualize the distribution of individual data points across categories.
##### Each observation is shown as a separate dot.
##### **Definition :** A Strip Plot displays all individual data points along a categorical axis.
##### **Example :**
| Department | Salary |
| ---------- | ------ |
| IT         | 50000  |
| IT         | 55000  |
| HR         | 35000  |
| HR         | 37000  |
| Sales      | 45000  |
| Sales      | 48000  |
##### **Why Use Strip Plot :**
##### Visualize Individual Data Points
##### Check Data Distribution
####### Compare Categories
##### Exploratory Data Analysis (EDA)
##### Small Dataset Analysis
##### Machine Learning Data Exploration
| Strip Plot                    | Scatter Plot                |
| ----------------------------- | --------------------------- |
| One axis categorical          | Both axes numerical         |
| Category comparison           | Relationship analysis       |
| Shows individual observations | Shows correlation           |
| Used with categories          | Used with numeric variables |
| Strip Plot          | Swarm Plot            |
| ------------------- | --------------------- |
| Dots may overlap    | No overlap            |
| Faster              | Slower                |
| Good for large data | Better for small data |
| Simpler             | Cleaner               |
##### **Syntax :** sns.stripplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    order=None,
    hue_order=None,
    jitter=True,
    dodge=False,
    orient=None,
    color=None,
    palette=None,
    size=5,
    alpha=None,
    linewidth=0,
    edgecolor='gray',
    legend='auto'
)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
df=pd.read_csv('Employee_Dataset.csv')
df
```

##### **X :** Categorical Variable
##### **Y :** Numerical Variable

```python
sns.stripplot(data=df,x='Department',y='Experience_Years')
plt.title('Strip Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Experience',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.5)
plt.tight_layout()
plt.show()
```

#### **hue :**
##### Creates separate colored points.

```python
sns.stripplot(data=df,x='Department',y='Experience_Years',hue='Gender')
plt.title('Strip Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Experience',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.5)
plt.tight_layout()
plt.show()
```
