### 3.Array with Built-in Methods
##### **3.Array with Built-in Methods :** Create a values of array and that array values can be performed with built in methods

##### **3.1 Sort() :** it sorts the array elements

```python
import numpy as np
arr = np.random.randint(1,10,5)
print(arr)
print(np.sort(arr))
```

##### **3.2 argsort() :** it will returns the sorted indexs of array elements

```python
arr = np.random.randint(1,10,5)
print(arr)
print(np.argsort(arr))
```

##### **3.3 min() :** Minimum array value will be returned

```python
arr = np.random.randint(1,10,5)
print(arr)
print(np.min(arr))
```

##### **3.4 argmin() :** Minimum array values index will be returned

```python
arr = np.random.randint(1,10,5)
print(arr)
print(np.argmin(arr))
```

##### **3.5 max() :** Maximum array value will be returned

```python
arr = np.random.randint(1,10,5)
print(arr)
print(np.max(arr))
```

##### **3.6 argmax() :** Maximum array values index will be returned

```python
arr = np.random.randint(1,10,5)
print(arr)
print(np.argmax(arr))
```

##### **3.7 sum() :** Sum of array value will be returned, Entire arrays sum will be returned

```python
arr = np.random.randint(1,10,5)
print(arr)
print(np.sum(arr))
```

##### **3.8 average() :** Average of array value will be returned, Entire arrays average will be returned

```python
arr = np.random.randint(1,10,5)
print(arr)
print(np.average(arr))
```

##### **3.9 mean() :** Mean of array value will be returned, Entire arrays mean will be returned
##### **Formula : Mean=Σx/n**
##### where:
##### Σx = Sum of all values
##### n = Number of values

```python
arr = np.random.randint(1,10,5)
print(arr)
print(np.mean(arr))
```

##### **3.10 median() :** Median of array value will be returned, Entire arrays median will be returned
##### **Formula : Median=((n+1)/2)th value**
##### Middle value after sorting

```python
arr = np.random.randint(1,10,5)
print(arr)
print(np.median(arr)) #Sorted array ka middle value , If even → average of two middle values
```

##### **3.11 sqrt() :** it will returns the square of the each elements.

```python
arr = np.random.randint(1,10,5)
print(arr)
print(np.sqrt(arr)) 
```

##### **3.12 var() :** Variance of array value will be returned, Entire arrays variance will be returned
##### **Formula :** variance=∑(x−μ)2​/N
##### where:
##### x = value
##### μ = mean
##### N = number of observations

```python
arr = np.random.randint(1,10,5) #Variance=n∑(x−mean)^2/total numbers
print(arr)
print(np.var(arr)) # mean=2.8 => (0.64+4.84+0.04+3.24+0.04)/5=1.76
```

##### **3.13 std() :** Standard of array value will be returned, Entire arrays Standar will be returned
##### **Formula :** σ=sqrt(σ2)

```python
arr = np.random.randint(1,10,5)
print(arr)
print(np.mean(arr))
print(np.var(arr)) 
print(np.std(arr)) #sqrt(var)
```

##### **mode() :** method will not support

##### **3.15 dot() :** Product of two array value will be returned, Entire two arrays product will be returned.
##### Why dot() Has No axis? Because dot() combines two arrays according to matrix multiplication rules.

```python
arr = np.random.randint(1,10,5)
print(arr)
print(np.dot(arr)) 
```

##### **1D Array**

```python
arr1 = np.random.randint(1,10,5)
arr2 = np.random.randint(1,10,5)
print(arr1)
print(arr2)
print(np.dot(arr1,arr2)) 
```

##### **2D Array**

```python
a = np.array([[1,2],
              [3,4]])
b = np.array([[5,6],
              [7,8]])
np.dot(a,b)
```

##### dot() is different. It performs matrix multiplication (dot product), not row-wise/column-wise aggregation.np.dot() does not have an axis parameter.

```python
arr1 = np.random.randint(1,10,5)
arr2 = np.random.randint(1,10,5)
print(arr1)
print(arr2)
print(np.dot(arr1,arr2),axis=0) 
```

| Feature | `np.prod()`                  | `np.dot()`                          |
| ------- | ---------------------------- | ----------------------------------- |
| Purpose | Product of all elements      | Dot Product / Matrix Multiplication |
| Inputs  | One array                    | Two arrays                          |
| Formula | a×b×c×d                      | a₁b₁+a₂b₂+a₃b₃                      |
| Output  | Single multiplication result | Sum of element-wise products        |
| Used In | Statistics                   | Linear Algebra, ML, AI              |

##### **Sum with axis=0 (multiple array):** **Row wise operation Column wise output**

##### **1D Array**

```python
arr1 = np.random.randint(1,10,5)
arr2 = np.random.randint(1,10,5)
print(arr1)
print(arr2)
print(np.sum((arr1,arr2),axis=0))
```

```python
arr1 = np.array([[1,2,3],
                [4,5,6]])
arr2 = np.array([[5,6,1],
                [7,4,2]])
print(arr1,'\n',arr2)
print(np.sum((arr1,arr2),axis=0))
```

##### **Sum with axis=1 (multiple array):** **Column wise operation Row wise output**
