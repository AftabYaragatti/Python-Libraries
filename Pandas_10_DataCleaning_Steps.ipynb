##### Out data contains

##### **format='mixed'** : The dates in this column may have different formats. Detect the format of each value individually.
##### **Example**
##### 2023-01-15
##### 2021/05/10
##### 15-08-2022
##### Without format='mixed', Pandas may expect all dates to follow the same pattern.
##### With format='mixed': Pandas converts:
##### 2023-01-15 → 2023-01-15
##### 2021/05/10 → 2021-05-10
##### 15-08-2022 → 2022-08-15

##### **errors='coerce'** : If a value cannot be converted into a valid date, don't throw an error. Replace it with NaT."
##### **Example**
##### df = pd.DataFrame({'Date': ['2023-01-15', 'invalid_date']})
##### **Without errors='coerce':**
##### pd.to_datetime(df['Date'])
##### ❌ Error: ValueError: Unknown datetime string format

##### **With errors='coerce':**
##### pd.to_datetime(df['Date'], errors='coerce')
##### Output:
##### 0   2023-01-15
##### 1          NaT
##### NaT = Not a Time (equivalent of NaN for dates)

##### **Other Values of errors :**
##### **errors='raise' (default)**
##### pd.to_datetime(df['Date'], errors='raise')
##### ❌ Stops execution and shows an error if any invalid date exists.

##### **errors='ignore'**
##### pd.to_datetime(df['Date'], errors='ignore')
##### Keeps invalid values unchanged.
##### Output:
##### 2023-01-15
##### invalid_date

##### pd.to_datetime(df['Date'],format='mixed',errors='coerce')
##### format='mixed' → Allows Pandas to parse multiple date formats in the same column.
##### errors='coerce' → Converts invalid dates into NaT instead of raising an error.

```python
df['Joining Date'] = pd.to_datetime(
    df['Joining Date'],
    format='mixed',
    errors='coerce'
)
```

##### **13 Outlier Detection and Treatment :**
##### Outliers are unusual values that are significantly different from other observations in the dataset.
##### Why Handle Outliers?
##### - Can distort statistical analysis.
##### - Can affect machine learning model performance.
##### - May indicate data entry errors.
##### - Methods to Detect Outliers:

##### **13.1 IQR (Interquartile Range) Method**
##### **Syntax : Q1 = df['Col_name'].quantile(0.25)**
##### **Syntax : Q2 = df['Col_name'].quantile(0.50)**
##### **Syntax : Q3 = df['Col_name'].quantile(0.75)**
##### **Formaula => IQR = Q3 - Q1**
##### **lower_bound = Q1 - 1.5 * IQR**
##### **upper_bound = Q3 + 1.5 * IQR**
##### **Common Indexing Methods in Pandas**
##### **1. loc (Label-based)**
##### Example : df.loc[df['Salary'] > upper_bound]
##### **2. iloc (Position-based)**
##### Example : df.iloc[0:5]
##### **Selects rows by position numbers**
##### **3. Fancy Indexing**
##### Example : df.iloc[[0, 2, 4]]
##### **Selects specific row positions**
##### **4. Boolean Indexing**
##### Example : df[df['Salary'] > upper_bound]
###### Example : df[(df['Salary'] < lower_bound) | (df['Salary'] > upper_bound)]
##### **outliers = df[(df['Salary'] < lower_bound) | (df['Salary'] > upper_bound)]**

```python
df = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Employee_Name': ['Aftab', 'Rahul', 'Priya', 'Anjali',
                      'Rohan', 'Kiran', 'Aman', 'Neha'],
    'Salary': [25000, 30000, 35000, 40000,
               45000, 50000, 55000, 200000]
})
df
```

##### **Step 1: Calculate Quartiles**

```python
Q1 = df['Salary'].quantile(0.25)
Q2 = df['Salary'].quantile(0.50)
Q3 = df['Salary'].quantile(0.75)

print("Q1 =", Q1)
print("Q2 =", Q2)
print("Q3 =", Q3)
```

##### **Step 2: Calculate IQR**

```python
IQR = Q3 - Q1
print("IQR =", IQR)
```

##### **Step 3: Calculate Lower and Upper Bounds**

```python
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)
```

```python
Step 4: Detect Outliers
Boolean Indexing
outliers = df[
    (df['Salary'] < lower_bound) |
    (df['Salary'] > upper_bound)
]
print(outliers)
```

##### **Using loc**

```python
df.loc[
    (df['Salary'] < lower_bound) |
    (df['Salary'] > upper_bound)
]
```

##### **Count Outliers**

```python
outliers.shape[0]
```

##### **Outlier Treatment Methods : => Method 1: Remove Outliers**

```python
df = df[
    (df['Salary'] >= lower_bound) &
    (df['Salary'] <= upper_bound)
]
df
```

##### **Method 2: Capping (Winsorization) :Replace extreme values with the upper/lower bound.**

```python
df['Salary'] = df['Salary'].clip(
    lower=lower_bound,
    upper=upper_bound
)
df
```

##### **Visualize Outliers Using Boxplot**

```python
import matplotlib.pyplot as plt
df.boxplot(column='Salary')
plt.show()
```

##### **13.2 Z-Score Method :**
##### Example : from scipy import stats
##### z_scores = stats.zscore(df['Salary'])
##### outliers = df[abs(z_scores) > 3]

##### The **Z-Score Method** identifies outliers based on how many standard deviations a value is away from the mean.
##### **Formula : [Z = \frac{X - Mean}{Standard\ Deviation}]**
##### Rule :
##### * `|Z| > 3` → Outlier
##### * `|Z| ≤ 3` → Normal Value

##### **Sample DataFrame**

```python
import pandas as pd
df = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Employee_Name': ['Aftab', 'Rahul', 'Priya', 'Anjali',
                      'Rohan', 'Kiran', 'Aman', 'Neha'],
    'Salary': [25000, 30000, 35000, 40000,
               45000, 50000, 55000, 200000]
})
df
```

##### **Step 1: Import SciPy**

```python
from scipy import stats
```

##### **Step 2: Calculate Z-Scores**

```python
z_scores = stats.zscore(df['Salary'])
z_scores
```

##### **Step 3: Detect Outliers**

```python
outliers = df[abs(z_scores) > 3]
outliers
```

##### **Compare IQR vs Z-Score**

| IQR Method                          | Z-Score Method                     |
| ----------------------------------- | ---------------------------------- |
| Uses Quartiles (Q1, Q3)             | Uses Mean & Standard Deviation     |
| Good for skewed data                | Best for normally distributed data |
| Less affected by extreme values     | Sensitive to extreme values        |
| Most commonly used in Data Cleaning | Common in Statistics & ML          |
