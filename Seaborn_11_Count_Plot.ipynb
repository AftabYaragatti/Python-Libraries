# **Seaborn Library for Data Analytics**

#### **What is a Count Plot :**
##### A Count Plot is used to show the frequency (count) of categorical data.
##### It counts how many times each category appears in the dataset.
##### **Definition :** A Count Plot displays the count of observations in each categorical category using bars.
##### **Example :**
| Gender |
| ------ |
| Male   |
| Female |
| Male   |
| Male   |
| Female |
##### **Count:** Male = 3 , Female = 2

##### **Why Use Count Plot :**
##### Category Frequency Analysis
##### Gender Distribution
##### Department Distribution
##### Customer Segmentation
##### Survey Analysis
##### Data Cleaning
##### Machine Learning EDA
##### **Syntax :** sns.countplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    order=None,
    hue_order=None,
    palette=None,
    color=None,
    saturation=0.75,
    width=0.8,
    dodge='auto',
    orient=None,
    legend='auto'
)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
df=pd.read_csv('Employee_Dataset.csv')
df
```

##### **X :** Displays categorical variable on X-axis.
##### **Y :** Creates a horizontal count plot.

```python
sns.countplot(data=df,x='Department')
plt.title('Counter Plot',fontsize=15,color='red',fontstyle='italic',fontfamily='Times New Roman')
plt.ylabel('Count (Frequency)',color='purple',fontsize=12)
plt.xlabel('Department',color='purple',fontsize=12)
plt.xticks(color='blue',rotation=90)
plt.yticks(color='blue',rotation=0)
plt.grid(True,axis='both',alpha=0.2)
plt.tight_layout()
plt.show()
```
