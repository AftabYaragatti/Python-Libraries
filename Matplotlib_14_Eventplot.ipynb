# **Matplotlib Library for Data Analytics**

### **Eventplot in Matplotlib**
##### An Event Plot is used to visualize the occurrence of events along a line or timeline.
##### **It answers:**
##### When did events occur?
##### How frequently did events occur?
##### Are events clustered together?
##### Are events spread out?
##### **What is an Event Plot?**
##### An Event Plot displays event positions as lines.
##### Each vertical line represents an event.

##### **Real-Life Examples :**
##### Website Visits
##### User Visits:
##### Stock Transactions
##### Each line = One trade.
##### Hospital Heartbeats (ECG)
##### Each line = Heartbeat.
##### Network Packets
##### Each line = Packet Arrival.
##### **Syntax :** plt.eventplot(
    positions,
    orientation='horizontal',
    lineoffsets=1,
    linelengths=1,
    linewidths=1,
    colors=None,
    alpha=None,
    linestyles='solid'
)

#### **positions :**
##### Most important parameter.
##### Specifies event locations.

```python
import matplotlib.pyplot as plt
events=[1,3,4,7,10,15]
plt.eventplot(events)
plt.show()
```

#### **orientation :**
##### Controls direction of event lines.(Vertical/Horizontal)

```python
import matplotlib.pyplot as plt
events=[1,3,4,7,10,15]
plt.eventplot(events,orientation='vertical')
plt.show()
```

```python
import matplotlib.pyplot as plt
events=[1,3,4,7,10,15]
plt.eventplot(events,orientation='horizontal')
plt.show()
```

#### **lineoffsets :**
##### Controls event row position.  
##### lineoffsets=5, Moves event line to position 5.

```python
import matplotlib.pyplot as plt
events=[1,3,4,7,10,15]
plt.eventplot(events,lineoffsets=5)
plt.show()
```

#### **linelengths & linewidths :**

```python
import matplotlib.pyplot as plt
events=[1,3,4,7,10,15]
plt.eventplot(events,lineoffsets=5,linelengths=3,linewidths=4)
plt.show()
```

#### **colors , alpha , linestyles :**
##### 
| Value     | Meaning  |
| --------- | -------- |
| 'solid'   | Solid    |
| 'dashed'  | Dashed   |
| 'dotted'  | Dotted   |
| 'dashdot' | Dash-dot |


```python
import matplotlib.pyplot as plt
events=[1,3,4,7,10,15]
plt.eventplot(events,lineoffsets=5,linelengths=3,linewidths=4,colors='purple',alpha=0.6,linestyle='dotted')
plt.show()
```

### **Multiple Eventplots :**

```python
import matplotlib.pyplot as plt
events=[[1,3,4,7,10,15],[5,6,3,2,6,8]]
plt.eventplot(events,lineoffsets=[5,2],linelengths=[3,2],linewidths=[5,2],colors=['purple','blue'],alpha=[0.6,0.8],linestyle=['dotted','dashed'])
plt.show()
```

```python
import matplotlib.pyplot as plt
events=[[1,3,4,7,10,15],[5,6,3,2,6,8]]
plt.eventplot(events,linelengths=[3,3],linewidths=[5,5],colors=['purple','blue'],alpha=[0.6,0.6],linestyle=['dotted','dashed'])
plt.show()
```

#### **Event Plot :** An Event Plot shows the positions where events occur.
##### **Example :** 
##### Think of it like:
##### Student attendance times
##### Website login times
##### Machine failures
##### Customer purchases
##### Bus arrival times
##### It shows when an event happened, not the event's value.

```python
import matplotlib.pyplot as plt
user1 = [2, 5, 8, 12, 15]
user2 = [1, 4, 7, 10, 14, 18]
user3 = [3, 6, 9, 11, 13, 16]
data = [user1, user2, user3]
plt.eventplot(data,colors=['green','red','blue'])
plt.title("Website Login Times",color='red',fontsize=20)
plt.xlabel("Hour of Day",fontsize=14,color='blue')
plt.ylabel("Users",fontsize=14,color='blue')
plt.grid(True)
plt.show()
```
