# **Pandas Library**
### **Conversion Methods in Pandas DataFrame()**

#### **WHAT ARE CONVERSION METHODS?**
##### Methods used to **convert a DataFrame into another format, structure, or data type**
##### 1. STRUCTURE CONVERSIONS (DataFrame → other pandas objects)
##### 2. FILE CONVERSIONS (DataFrame → external formats)
##### 3. TYPE CONVERSIONS (Data types)
##### 4. INDEX / AXIS CONVERSIONS
##### 5. SHAPE / FORMAT CONVERSIONS
##### 6. COPY / DUPLICATION

#### **1. STRUCTURE CONVERSIONS (DataFrame → other pandas objects)**
##### 1. df.to_numpy() : Convert DataFrame → NumPy array
##### 2. df.values(older) : Same as to_numpy()
##### 3. df.squeeze() : Convert DataFrame → Series (if possible)
##### 4. df.to_dict() : Convert DataFrame → dictionary
##### 5. df.to_records() : Convert → structured NumPy array
##### 6. df.to_string() : Convert → string format

#### **2. FILE CONVERSIONS (DataFrame → external formats)**
##### 7. df.to_csv() : df.to_csv(path, index=True), Save as CSV
##### 8. df.to_excel() : df.to_excel(path)
##### 9. df.to_json() : df.to_json(path_or_buf=None, orient=None)
##### 10. df.to_html() : Convert → HTML table
##### 11. df.to_sql() : df.to_sql(name, con), Store in database
##### 12. df.to_markdown() : Convert → markdown table

#### **3. TYPE CONVERSIONS (Data types)**
##### 13. df.astype() : Convert column data types, df.astype(dtype)
##### 14. df.convert_dtypes() : Automatically convert to best dtype
##### 15. pd.to_numeric() : pd.to_numeric(df['col'])
##### 16. pd.to_datetime() : pd.to_datetime(df['date'])

#### **4. INDEX / AXIS CONVERSIONS**
##### 18. df.reset_index() : Convert index → column
##### 19. df.set_index() : Convert column → index
##### 20. df.rename_axis() : Convert axis labels

#### **5. SHAPE / FORMAT CONVERSIONS**
##### 21. df.stack() : Columns → rows
##### 22. df.unstack() : Rows → columns
##### 23. df.melt() : Wide → long format
##### 24. df.pivot() : Long → wide format

#### **6. COPY / DUPLICATION**
##### 26. df.copy() : Create copy

##### **1. STRUCTURE CONVERSIONS (DataFrame → other pandas objects)**

##### **1. df.to_numpy() :** Convert DataFrame → NumPy array
##### **Syntax : df.to_numpy(dtype=None, copy=False)**

```python
import pandas as pd
import numpy as np
df=pd.DataFrame(data={'Student_ID':[101,102,103,104,105],'Names':['Aftab','Smith','Scott','Bob','Jhon'],'Marks1':np.arange(70,95,5),
                      'Marks2':np.arange(50,75,5),'Marks3':np.arange(40,65,5),'Marks4':np.arange(75,100,5),'Marks5':np.arange(65,90,5),
                      'Marks6':np.arange(70,95,5)},index=np.arange(1,6))
df
```

```python
df.to_numpy()
```

##### IF we want to change the type of data we can change by using dtype attribute.

```python
df.to_numpy(dtype='int')
```

##### Note : We can't change the string into int, we can change int to str

```python
df.to_numpy(dtype='str')
```

```python
arr2=df.to_numpy(copy=True)
```

```python
arr2
```

```python
arr2[0,0]=100
```
