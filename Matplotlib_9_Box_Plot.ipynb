# **Matplotlib Library for Data Analytics**

### **Box Plot in Matplotlib**

##### A Box Plot (also called Box-and-Whisker Plot) is one of the most important charts in Data Analytics, Statistics, Data Science, Machine Learning, and Exploratory Data Analysis (EDA).
##### **It helps us understand:**
##### Data Distribution
##### Median
##### Quartiles
##### Spread of Data
##### Outliers
##### Skewness
##### **What is a Box Plot?**
##### A Box Plot is a graphical representation of the Five Number Summary:
##### Minimum
##### First Quartile (Q1)
##### Median (Q2)
##### Third Quartile (Q3)
##### Maximum

##### **Why Use Box Plot?**
##### Detect Outliers : 10 12 15 18 20 22 25 200
##### **Here:** 200 is an outlier.
##### **Example :**
##### Consider: data=[10,20,30,40,50,60,70,80,90]
##### **Minimum :** 10
##### **Q1 (25%) :** 30
##### **Median (50%) :** 50
##### **Q3 (75%) :** 70
##### **Maximum :** 90
##### **Box Plot Syntax :** plt.boxplot(
    x,
    notch=False,
    vert=True,
    patch_artist=False,
    widths=None,
    labels=None,
    showmeans=False,
    meanline=False,
    showfliers=True,
    whis=1.5,
    medianprops=None,
    boxprops=None,
    whiskerprops=None,
    capprops=None,
    flierprops=None )

```python
import matplotlib.pyplot as plt
marks=[92,34,56,97,1,23,45,76,89,120]
plt.boxplot(marks)
plt.grid(True)
plt.show()
```

#### **medianprops :**
##### Median line style.
##### medianprops=dict(color='red', linewidth=2 )

```python
import matplotlib.pyplot as plt
marks=[92,34,56,97,1,23,45,76,89,120]
plt.boxplot(marks,showmeans=True,meanline=True,medianprops=dict(color='blue', linewidth=2 ))
plt.grid(True)
plt.show()
```

#### **boxprops :** 
##### Box style
##### boxprops=dict(color='blue',linewidth=2,linestyle='--')

```python
import matplotlib.pyplot as plt
marks=[92,34,56,97,1,23,45,76,89,120]
plt.boxplot(marks,showmeans=True,meanline=True,medianprops=dict(color='blue', linewidth=2),boxprops=dict(color='purple',linewidth=2,linestyle='--'))
plt.grid(True)
plt.show()
```

#### **whiskerprops :**
##### Whiskers style.
##### whiskerprops=dict(color='blue',linewidth=2,linestyle='--')

```python
import matplotlib.pyplot as plt
marks=[92,34,56,97,1,23,45,76,89,120]
plt.boxplot(marks,showmeans=True,meanline=True,medianprops=dict(color='blue', linewidth=2),boxprops=dict(color='purple'),whiskerprops=dict(color='green',linewidth=2))
plt.grid(True)
plt.show()
```

#### **capprops :**
##### Outlier style.
##### capprops=dict(color='blue',linewidth=2,linestyle='--')

```python
import matplotlib.pyplot as plt
marks=[92,34,56,97,1,23,45,76,89,120]
plt.boxplot(marks,showmeans=True,meanline=True,medianprops=dict(color='blue', linewidth=2),boxprops=dict(color='purple',linewidth=2),whiskerprops=dict(color='green',linewidth=2),capprops=dict(color='purple',linewidth=3))
plt.grid(True)
plt.show()
```

#### **flierprops :**
##### Outlier style.
##### flierprops=dict(marker='o')

```python
import matplotlib.pyplot as plt
marks=[92,34,56,97,1,23,45,76,89,150]
plt.boxplot(marks,flierprops=dict(marker='*'))
plt.grid(True)
plt.show()
```

### **Multiple Box Plot :**

```python
import matplotlib.pyplot as plt
marks1=[92,34,56,97,1,23,45,76,89,120]
marks2=[67,45,34,76,76,89,23,73,43,167]
marks3=[76,23,87,9,43,97,76,54,34,145]
plt.boxplot([marks1,marks2,marks3],showmeans=True,meanline=True,medianprops=dict(color='blue', linewidth=2),boxprops=dict(color='purple',linewidth=2),whiskerprops=dict(color='green',linewidth=2),capprops=dict(color='purple',linewidth=3))
plt.grid(True)
plt.show()
```
